```python
# Create a fictional internal product catalogue

N_PRODUCTS = 3000

categories = [
    "Womenswear",
    "Menswear",
    "Kidswear",
    "Footwear",
    "Accessories",
    "Home"
]

category_weights = [0.28, 0.20, 0.15, 0.14, 0.10, 0.13]

brand_names = [
    "Northline",
    "Harbour & Pine",
    "Mosaic House",
    "Everwell",
    "Juniper Lane",
    "Cedar & Coast",
    "Aster Row",
    "Fieldstone",
    "Lumen Studio",
    "Kindred Goods",
    "Oak & Vale",
    "Solace"
]

sizes_by_category = {
    "Womenswear": ["XS", "S", "M", "L", "XL"],
    "Menswear": ["S", "M", "L", "XL", "XXL"],
    "Kidswear": ["2Y", "4Y", "6Y", "8Y", "10Y", "12Y"],
    "Footwear": ["36", "37", "38", "39", "40", "41", "42", "43", "44"],
    "Accessories": ["One Size"],
    "Home": ["Small", "Medium", "Large"]
}

weight_ranges = {
    "Womenswear": (0.15, 1.50),
    "Menswear": (0.20, 1.80),
    "Kidswear": (0.10, 1.20),
    "Footwear": (0.40, 2.00),
    "Accessories": (0.05, 0.80),
    "Home": (0.30, 8.00)
}

product_rows = []

project_start_date = pd.Timestamp(PROJECT_START)
project_end_date = pd.Timestamp(PROJECT_END)

for product_number in range(1, N_PRODUCTS + 1):
    category = rng.choice(categories, p=category_weights)
    brand_name = rng.choice(brand_names)
    size = rng.choice(sizes_by_category[category])
    
    min_weight, max_weight = weight_ranges[category]
    weight_kg = rng.uniform(min_weight, max_weight)
    
    # Product costs are synthetic and expressed in GBP
    unit_cost_gbp = np.clip(
        rng.lognormal(mean=2.5, sigma=0.65),
        3.50,
        180.00
    )
    
    # Most products remain active for the full project period
    active_from = project_start_date + pd.Timedelta(
        days=int(rng.integers(0, 240))
    )
    
    if rng.random() < 0.08:
        active_to = project_end_date - pd.Timedelta(
            days=int(rng.integers(15, 120))
        )
    else:
        active_to = project_end_date
    
    product_rows.append({
        "internal_sku": f"SKU_{product_number:06d}",
        "parent_product_id": f"PARENT_{int(rng.integers(1, 850)):05d}",
        "category": category,
        "brand_name": brand_name,
        "brand_type": rng.choice(
            ["owned", "licensed", "third_party"],
            p=[0.45, 0.15, 0.40]
        ),
        "size": size,
        "colour": rng.choice([
            "Black",
            "White",
            "Navy",
            "Stone",
            "Green",
            "Blue",
            "Red",
            "Natural"
        ]),
        "unit_cost_gbp": round(float(unit_cost_gbp), 2),
        "weight_kg": round(float(weight_kg), 3),
        "active_from": active_from,
        "active_to": active_to
    })

product_ref = pd.DataFrame(product_rows)

print("Products created:", len(product_ref))
print("Columns:")
print(list(product_ref.columns))

display(product_ref.head())
```

```python
# Create fictional mappings between internal products and partner SKUs

mapping_rows = []

partner_listing_probabilities = {
    "CH_A": 0.75,
    "CH_B": 0.65,
    "CH_C": 0.50,
    "CH_D": 0.40
}

for _, product in product_ref.iterrows():
    for partner_id, listing_probability in partner_listing_probabilities.items():
        if rng.random() < listing_probability:
            mapping_rows.append({
                "partner_id": partner_id,
                "internal_sku": product["internal_sku"],
                "partner_sku": f"{partner_id}_{product['internal_sku']}",
                "mapping_method": rng.choice(
                    ["automated", "manual", "partner_confirmed"],
                    p=[0.70, 0.15, 0.15]
                ),
                "mapping_status": "mapped",
                "effective_from": product["active_from"],
                "effective_to": product["active_to"]
            })

partner_product_map = pd.DataFrame(mapping_rows)

print("Partner-product mappings created:", len(partner_product_map))
print("\nMappings by partner:")
display(
    partner_product_map
    .groupby("partner_id")
    .size()
    .reset_index(name="mapping_count")
)
```

## Generate an unfamiliar partner listing export

This file represents weekly product-listing snapshots received from fictional external selling channels.

The raw file intentionally uses source-style column names and mixed formats. We will not clean it during generation. Cleaning and investigation will happen in later steps.

```python
# Prepare weekly listing snapshots from the fictional partner-product mappings

snapshot_dates = pd.date_range(
    start=PROJECT_START,
    end=PROJECT_END,
    freq="7D"
)

listing_base = (
    partner_product_map
    .merge(
        product_ref[
            [
                "internal_sku",
                "parent_product_id",
                "category",
                "brand_name",
                "unit_cost_gbp",
                "active_from",
                "active_to"
            ]
        ],
        on="internal_sku",
        how="left"
    )
    .merge(
        partner_ref[
            [
                "partner_id",
                "active_from"
            ]
        ].rename(
            columns={"active_from": "partner_active_from"}
        ),
        on="partner_id",
        how="left"
    )
)

listing_snapshots = []

for snapshot_date in snapshot_dates:
    snapshot = listing_base[
        (listing_base["active_from"] <= snapshot_date)
        & (listing_base["active_to"] >= snapshot_date)
        & (listing_base["partner_active_from"] <= snapshot_date)
    ].copy()

    snapshot["snapshot_date"] = snapshot_date
    listing_snapshots.append(snapshot)

listing_snapshots = pd.concat(
    listing_snapshots,
    ignore_index=True
)

print("Weekly snapshot dates:", len(snapshot_dates))
print("Base listing snapshot rows:", len(listing_snapshots))
```

```python
# Create source-style fields for the external listing export

n_rows = len(listing_snapshots)

# Assign a marketplace country to each listing snapshot
country_codes = country_ref["country_code"].to_numpy()

listing_snapshots["marketplace_country_clean"] = rng.choice(
    country_codes,
    size=n_rows,
    p=[0.24, 0.08, 0.16, 0.13, 0.09, 0.05, 0.08, 0.07, 0.06, 0.04]
)

# Synthetic price in GBP before conversion to the displayed source currency
markup_factor = rng.uniform(1.8, 3.4, size=n_rows)

listing_snapshots["synthetic_price_gbp"] = (
    listing_snapshots["unit_cost_gbp"] * markup_factor
).round(2)

# Add seasonality and modest price movement
month_number = listing_snapshots["snapshot_date"].dt.month

seasonal_multiplier = np.where(
    month_number.isin([11, 12]),
    1.08,
    np.where(
        month_number.isin([1, 7]),
        0.94,
        1.00
    )
)

listing_snapshots["synthetic_price_gbp"] = (
    listing_snapshots["synthetic_price_gbp"]
    * seasonal_multiplier
    * rng.uniform(0.97, 1.03, size=n_rows)
).round(2)

# Synthetic stock position
base_stock = rng.gamma(
    shape=2.2,
    scale=5.5,
    size=n_rows
)

listing_snapshots["reported_stock_clean"] = np.maximum(
    np.round(base_stock).astype(int),
    0
)

# Create natural zero-stock situations
zero_stock_mask = rng.random(n_rows) < 0.08
listing_snapshots.loc[
    zero_stock_mask,
    "reported_stock_clean"
] = 0

# Create listing states using stock, lifecycle, and source behaviour
listing_states = np.full(n_rows, "LIVE", dtype=object)

paused_mask = (
    (rng.random(n_rows) < 0.035)
    & (listing_snapshots["reported_stock_clean"] > 0)
)

preorder_mask = (
    (rng.random(n_rows) < 0.018)
    & (listing_snapshots["reported_stock_clean"] == 0)
)

offline_mask = (
    (listing_snapshots["reported_stock_clean"] == 0)
    | (rng.random(n_rows) < 0.025)
)

listing_states[paused_mask] = "PAUSED"
listing_states[preorder_mask] = "PREORDER"
listing_states[offline_mask] = "OFFLINE"

listing_snapshots["listing_state_clean"] = listing_states

print(
    listing_snapshots["listing_state_clean"]
    .value_counts()
)
```

```python
# Repair cell: create all source-style fields in one step

n_rows = len(listing_snapshots)

# Use a separate random generator for this stage
listing_rng = np.random.default_rng(RANDOM_SEED + 16)

def format_mixed_date(value, format_number):
    if format_number == 0:
        return value.strftime("%Y-%m-%d")
    elif format_number == 1:
        return value.strftime("%d/%m/%Y")
    else:
        return value.strftime("%Y-%m-%d %H:%M:%S")


def format_currency(currency_code, format_number):
    aliases = {
        "GBP": ["GBP", "£", "Pound Sterling"],
        "EUR": ["EUR", "€", "Euro"],
        "SEK": ["SEK", "kr", "Swedish Krona"],
        "DKK": ["DKK", "kr", "Danish Krone"]
    }

    return aliases[currency_code][format_number]


country_currency_lookup = (
    country_ref
    .set_index("country_code")["currency_code"]
    .to_dict()
)

# External partner identifier
listing_snapshots["source_partner_code"] = np.where(
    listing_rng.random(n_rows) < 0.035,
    " " + listing_snapshots["partner_id"].astype(str) + " ",
    listing_snapshots["partner_id"].astype(str)
)

# External country labels
listing_snapshots["source_market"] = (
    listing_snapshots["marketplace_country_clean"]
    .map({
        "GB": "UK",
        "IE": "IE",
        "DE": "DE",
        "FR": "FR",
        "NL": "NL",
        "BE": "BE",
        "ES": "ES",
        "IT": "IT",
        "SE": "SWE",
        "DK": "DK"
    })
)

# External product identifier
listing_snapshots["source_item_code"] = (
    listing_snapshots["partner_sku"].astype(str)
)

identifier_whitespace_mask = listing_rng.random(n_rows) < 0.025

listing_snapshots.loc[
    identifier_whitespace_mask,
    "source_item_code"
] = (
    " "
    + listing_snapshots.loc[
        identifier_whitespace_mask,
        "source_item_code"
    ]
    + " "
)

# External currency
listing_snapshots["source_currency_clean"] = (
    listing_snapshots["marketplace_country_clean"]
    .map(country_currency_lookup)
)

# Synthetic exchange rates
synthetic_fx_rates = {
    "GBP": 1.00,
    "EUR": 1.17,
    "SEK": 0.087,
    "DKK": 0.134
}

listing_snapshots["offer_price_clean"] = (
    listing_snapshots["synthetic_price_gbp"]
    / listing_snapshots["source_currency_clean"]
    .map(synthetic_fx_rates)
).round(2)

currency_format_number = listing_rng.integers(
    0,
    3,
    size=n_rows
)

listing_snapshots["source_currency"] = [
    format_currency(
        currency,
        int(format_number)
    )
    for currency, format_number in zip(
        listing_snapshots["source_currency_clean"],
        currency_format_number
    )
]

# Mixed date formats
date_format_number = listing_rng.integers(
    0,
    3,
    size=n_rows
)

listing_snapshots["source_snapshot_date"] = [
    format_mixed_date(
        date_value,
        int(format_number)
    )
    for date_value, format_number in zip(
        listing_snapshots["snapshot_date"],
        date_format_number
    )
]

# Inventory update timestamp
inventory_update_offset_hours = listing_rng.integers(
    0,
    96,
    size=n_rows
)

listing_snapshots["inventory_update_timestamp"] = (
    listing_snapshots["snapshot_date"]
    + pd.to_timedelta(
        inventory_update_offset_hours,
        unit="h"
    )
)

inventory_date_format_number = listing_rng.integers(
    0,
    3,
    size=n_rows
)

listing_snapshots["source_inventory_update"] = [
    format_mixed_date(
        timestamp_value,
        int(format_number)
    )
    for timestamp_value, format_number in zip(
        listing_snapshots["inventory_update_timestamp"],
        inventory_date_format_number
    )
]

# Source listing fields
listing_snapshots["source_listing_state"] = (
    listing_snapshots["listing_state_clean"]
)

listing_snapshots["source_stock"] = (
    listing_snapshots["reported_stock_clean"]
)

listing_snapshots["source_price"] = (
    listing_snapshots["offer_price_clean"]
)

listing_snapshots["source_category"] = (
    listing_snapshots["category"]
)

listing_snapshots["source_title"] = (
    listing_snapshots["brand_name"]
    + " "
    + listing_snapshots["category"]
    + " "
    + listing_snapshots["internal_sku"]
)

listing_snapshots["source_image_url"] = (
    "https://fictional.example/images/"
    + listing_snapshots["source_item_code"].astype(str).str.strip()
    + ".jpg"
)

# Incomplete source attributes
title_missing_mask = listing_rng.random(n_rows) < 0.012
image_missing_mask = listing_rng.random(n_rows) < 0.018
category_missing_mask = listing_rng.random(n_rows) < 0.010

listing_snapshots.loc[
    title_missing_mask,
    "source_title"
] = None

listing_snapshots.loc[
    image_missing_mask,
    "source_image_url"
] = None

listing_snapshots.loc[
    category_missing_mask,
    "source_category"
] = None

# Category formatting variation
category_format_mask = listing_rng.random(n_rows) < 0.035

listing_snapshots.loc[
    category_format_mask,
    "source_category"
] = (
    listing_snapshots.loc[
        category_format_mask,
        "source_category"
    ]
    .astype("string")
    .str.lower()
    .str.strip()
)

# File lineage and ingestion timestamp
listing_snapshots["source_file"] = (
    "partner_listing_"
    + listing_snapshots["partner_id"].astype(str)
    + "_"
    + listing_snapshots["snapshot_date"]
    .dt.strftime("%Y%m%d")
    + ".csv"
)

listing_snapshots["ingested_at"] = (
    listing_snapshots["snapshot_date"]
    + pd.to_timedelta(
        listing_rng.integers(
            1,
            72,
            size=n_rows
        ),
        unit="h"
    )
)

print("Source-style fields created successfully.")
print(f"Current row count: {len(listing_snapshots):,}")
```

## Generate an unfamiliar inventory export

This file represents inventory snapshots received from fictional external partners or fulfilment sources.

The inventory export is intentionally separate from the listing export. Later, we will compare the two sources to investigate stale stock, availability disagreements, and possible mapping issues.

```python
# Create a clean base for the fictional inventory source

inventory_rng = np.random.default_rng(RANDOM_SEED + 40)

inventory_base = (
    listing_snapshots[
        [
            "partner_id",
            "partner_sku",
            "marketplace_country_clean",
            "snapshot_date",
            "reported_stock_clean"
        ]
    ]
    .drop_duplicates(
        subset=[
            "partner_id",
            "partner_sku",
            "marketplace_country_clean",
            "snapshot_date"
        ]
    )
    .copy()
)

inventory_base = inventory_base.reset_index(drop=True)

n_inventory_rows = len(inventory_base)

print("Inventory base rows:", f"{n_inventory_rows:,}")
display(inventory_base.head())
```

```python
# Create synthetic inventory quantities

# Add small source-specific differences to the listing stock values
stock_adjustment = inventory_rng.choice(
    [-3, -2, -1, 0, 0, 0, 1, 2, 3],
    size=n_inventory_rows
)

inventory_base["available_units_clean"] = np.maximum(
    inventory_base["reported_stock_clean"] + stock_adjustment,
    0
)

# Reserved stock is usually a fraction of available stock
reservation_fraction = inventory_rng.uniform(
    0.00,
    0.35,
    size=n_inventory_rows
)

inventory_base["reserved_units_clean"] = np.minimum(
    np.floor(
        inventory_base["available_units_clean"]
        * reservation_fraction
    ).astype(int),
    inventory_base["available_units_clean"]
)

inventory_base["inbound_units_clean"] = inventory_rng.poisson(
    lam=2.5,
    size=n_inventory_rows
)

# Create fulfilment locations
inventory_base["warehouse_clean"] = inventory_rng.choice(
    [
        "WH_NORTH",
        "WH_CENTRAL",
        "WH_SOUTH",
        "WH_EUROPE"
    ],
    size=n_inventory_rows,
    p=[0.25, 0.40, 0.20, 0.15]
)

print("Inventory quantities created.")
display(
    inventory_base[
        [
            "available_units_clean",
            "reserved_units_clean",
            "inbound_units_clean",
            "warehouse_clean"
        ]
    ].describe()
)
```

```python
# Convert clean inventory values into external source formats

def format_inventory_date(value, format_number):
    if format_number == 0:
        return value.strftime("%Y-%m-%d")
    elif format_number == 1:
        return value.strftime("%d/%m/%Y")
    else:
        return value.strftime("%Y-%m-%d %H:%M:%S")


inventory_export = pd.DataFrame()

# Partner identifier with occasional whitespace
inventory_export["partner_code"] = np.where(
    inventory_rng.random(n_inventory_rows) < 0.025,
    " " + inventory_base["partner_id"].astype(str) + " ",
    inventory_base["partner_id"].astype(str)
)

# Partner SKU with occasional whitespace
inventory_export["item_code"] = inventory_base[
    "partner_sku"
].astype(str)

item_whitespace_mask = inventory_rng.random(
    n_inventory_rows
) < 0.022

inventory_export.loc[
    item_whitespace_mask,
    "item_code"
] = (
    " "
    + inventory_export.loc[
        item_whitespace_mask,
        "item_code"
    ]
    + " "
)

# Country labels differ from the country reference table
inventory_export["market"] = (
    inventory_base["marketplace_country_clean"]
    .map({
        "GB": "UK",
        "IE": "IE",
        "DE": "DE",
        "FR": "FR",
        "NL": "NLD",
        "BE": "BE",
        "ES": "ESP",
        "IT": "IT",
        "SE": "SWE",
        "DK": "DK"
    })
)

inventory_export["warehouse"] = inventory_base[
    "warehouse_clean"
]

# Numeric source fields
inventory_export["available_qty"] = inventory_base[
    "available_units_clean"
]

inventory_export["reserved_qty"] = inventory_base[
    "reserved_units_clean"
]

inventory_export["inbound_qty"] = inventory_base[
    "inbound_units_clean"
]

# Inventory update timestamps
inventory_update_hours = inventory_rng.integers(
    2,
    120,
    size=n_inventory_rows
)

inventory_export["inventory_timestamp_clean"] = (
    inventory_base["snapshot_date"]
    + pd.to_timedelta(
        inventory_update_hours,
        unit="h"
    )
)

inventory_date_formats = inventory_rng.integers(
    0,
    3,
    size=n_inventory_rows
)

inventory_export["inventory_timestamp"] = [
    format_inventory_date(
        timestamp_value,
        int(format_number)
    )
    for timestamp_value, format_number in zip(
        inventory_export["inventory_timestamp_clean"],
        inventory_date_formats
    )
]

# File lineage
inventory_export["source_file"] = (
    "inventory_"
    + inventory_export["partner_code"].astype(str).str.strip()
    + "_"
    + inventory_base["snapshot_date"].dt.strftime("%Y%m%d")
    + ".csv"
)

inventory_export["ingested_at"] = (
    inventory_export["inventory_timestamp_clean"]
    + pd.to_timedelta(
        inventory_rng.integers(
            1,
            48,
            size=n_inventory_rows
        ),
        unit="h"
    )
)

print("Source-specific inventory fields created.")
display(inventory_export.head())
```

## Generate an unfamiliar order-header export

This file represents one row per order received from fictional third-party selling channels.

Order-line data will be generated separately so that we can later test whether order headers and lines reconcile correctly.

```python
# Create synthetic order dates with realistic seasonal behaviour

order_rng = np.random.default_rng(RANDOM_SEED + 50)

N_ORDERS = 75000

order_dates = pd.date_range(
    start=PROJECT_START,
    end=PROJECT_END,
    freq="D"
)

# Start with a baseline probability for each date
daily_weights = np.ones(len(order_dates), dtype=float)

# Weekends receive slightly higher online order volume
weekend_mask = order_dates.dayofweek >= 5
daily_weights[weekend_mask] *= 1.15

# November and December receive higher volume
holiday_mask = order_dates.month.isin([11, 12])
daily_weights[holiday_mask] *= 1.35

# January is quieter after the holiday period
january_mask = order_dates.month == 1
daily_weights[january_mask] *= 0.82

daily_probabilities = daily_weights / daily_weights.sum()

sampled_order_dates = order_rng.choice(
    order_dates.to_numpy(),
    size=N_ORDERS,
    p=daily_probabilities
)

sampled_order_dates = pd.to_datetime(
    sampled_order_dates
)

order_header = pd.DataFrame({
    "order_date_clean": sampled_order_dates
})

# Add a random order time during the day
order_header["order_timestamp_clean"] = (
    order_header["order_date_clean"]
    + pd.to_timedelta(
        order_rng.integers(
            0,
            24 * 60 * 60,
            size=N_ORDERS
        ),
        unit="s"
    )
)

print("Orders generated:", len(order_header))
print(
    "Date range:",
    order_header["order_timestamp_clean"].min(),
    "to",
    order_header["order_timestamp_clean"].max()
)
```

```python
# Assign fictional partners and countries to each order

partner_choices = []
country_choices = []

partner_country_options = {
    "CH_A": {
        "countries": ["GB", "IE", "DE"],
        "weights": [0.60, 0.20, 0.20]
    },
    "CH_B": {
        "countries": ["DE", "FR", "NL", "BE"],
        "weights": [0.35, 0.30, 0.20, 0.15]
    },
    "CH_C": {
        "countries": ["FR", "ES", "IT", "DE"],
        "weights": [0.25, 0.30, 0.25, 0.20]
    },
    "CH_D": {
        "countries": ["SE", "DK", "DE"],
        "weights": [0.45, 0.30, 0.25]
    }
}

for order_date in order_header["order_timestamp_clean"]:
    if order_date < pd.Timestamp("2025-03-01"):
        eligible_partners = ["CH_A", "CH_B"]
        eligible_weights = [0.62, 0.38]
    elif order_date < pd.Timestamp("2025-06-01"):
        eligible_partners = ["CH_A", "CH_B", "CH_C"]
        eligible_weights = [0.48, 0.32, 0.20]
    else:
        eligible_partners = ["CH_A", "CH_B", "CH_C", "CH_D"]
        eligible_weights = [0.40, 0.28, 0.20, 0.12]

    selected_partner = order_rng.choice(
        eligible_partners,
        p=eligible_weights
    )

    partner_choices.append(selected_partner)

    country_config = partner_country_options[selected_partner]

    selected_country = order_rng.choice(
        country_config["countries"],
        p=country_config["weights"]
    )

    country_choices.append(selected_country)

order_header["partner_id_clean"] = partner_choices
order_header["country_code_clean"] = country_choices

print("Orders by partner:")
display(
    order_header["partner_id_clean"]
    .value_counts()
    .sort_index()
    .to_frame("order_count")
)

print("Orders by country:")
display(
    order_header["country_code_clean"]
    .value_counts()
    .sort_index()
    .to_frame("order_count")
)
```

```python
# Create synthetic order amounts and statuses

# Basket values are positively skewed
order_header["item_amount_clean"] = np.clip(
    order_rng.lognormal(
        mean=3.65,
        sigma=0.75,
        size=N_ORDERS
    ),
    8.00,
    950.00
).round(2)

order_header["shipping_amount_clean"] = np.select(
    [
        order_header["item_amount_clean"] >= 100,
        order_header["item_amount_clean"] >= 50
    ],
    [
        0.00,
        3.99
    ],
    default=5.99
).round(2)

order_header["discount_amount_clean"] = np.where(
    order_rng.random(N_ORDERS) < 0.28,
    (
        order_header["item_amount_clean"]
        * order_rng.uniform(0.05, 0.25, size=N_ORDERS)
    ).round(2),
    0.00
)

country_tax_lookup = (
    country_ref
    .set_index("country_code")["synthetic_tax_rate"]
    .to_dict()
)

order_header["synthetic_tax_rate"] = (
    order_header["country_code_clean"]
    .map(country_tax_lookup)
)

order_header["tax_amount_clean"] = (
    (
        order_header["item_amount_clean"]
        - order_header["discount_amount_clean"]
    )
    * order_header["synthetic_tax_rate"]
).round(2)

# More recent orders are more likely to remain pending
age_in_days = (
    pd.Timestamp(PROJECT_END)
    - order_header["order_timestamp_clean"].dt.normalize()
).dt.days

pending_probability = np.where(
    age_in_days < 10,
    0.10,
    0.015
)

random_status_value = order_rng.random(N_ORDERS)

order_status = np.select(
    [
        random_status_value < 0.055,
        random_status_value < 0.095,
        random_status_value < 0.16,
        random_status_value < 0.22,
        random_status_value < 0.22 + pending_probability
    ],
    [
        "cancelled",
        "returned",
        "refunded",
        "shipped",
        "pending"
    ],
    default="delivered"
)

order_header["order_status_clean"] = order_status

print("Order statuses:")
display(
    order_header["order_status_clean"]
    .value_counts()
    .to_frame("order_count")
)
```

```python
# Convert order-header values into external source formats

def format_order_timestamp(value, format_number):
    if format_number == 0:
        return value.strftime("%Y-%m-%d %H:%M:%S")
    elif format_number == 1:
        return value.strftime("%d/%m/%Y %H:%M")
    else:
        return value.strftime("%Y-%m-%dT%H:%M:%S")


def order_currency_alias(currency_code, format_number):
    aliases = {
        "GBP": ["GBP", "£", "Pound Sterling"],
        "EUR": ["EUR", "€", "Euro"],
        "SEK": ["SEK", "kr", "Swedish Krona"],
        "DKK": ["DKK", "kr", "Danish Krone"]
    }

    return aliases[currency_code][format_number]


country_currency_lookup = (
    country_ref
    .set_index("country_code")["currency_code"]
    .to_dict()
)

order_header["source_order_id"] = [
    f"ORD-{number:09d}"
    for number in range(1, N_ORDERS + 1)
]

order_header["source_partner_code"] = np.where(
    order_rng.random(N_ORDERS) < 0.025,
    " " + order_header["partner_id_clean"] + " ",
    order_header["partner_id_clean"]
)

order_header["source_market"] = (
    order_header["country_code_clean"]
    .map({
        "GB": "UK",
        "IE": "IE",
        "DE": "DE",
        "FR": "FR",
        "NL": "NLD",
        "BE": "BE",
        "ES": "ESP",
        "IT": "IT",
        "SE": "SWE",
        "DK": "DK"
    })
)

order_header["currency_clean"] = (
    order_header["country_code_clean"]
    .map(country_currency_lookup)
)

currency_format_number = order_rng.integers(
    0,
    3,
    size=N_ORDERS
)

order_header["source_currency"] = [
    order_currency_alias(
        currency,
        int(format_number)
    )
    for currency, format_number in zip(
        order_header["currency_clean"],
        currency_format_number
    )
]

timestamp_format_number = order_rng.integers(
    0,
    3,
    size=N_ORDERS
)

order_header["source_created_at"] = [
    format_order_timestamp(
        timestamp,
        int(format_number)
    )
    for timestamp, format_number in zip(
        order_header["order_timestamp_clean"],
        timestamp_format_number
    )
]

order_header["source_status"] = (
    order_header["order_status_clean"]
    .map({
        "delivered": "COMPLETED",
        "shipped": "DISPATCHED",
        "pending": "OPEN",
        "cancelled": "CANCELLED",
        "returned": "RETURNED",
        "refunded": "REFUNDED"
    })
)

order_header["source_item_amount"] = (
    order_header["item_amount_clean"]
)

order_header["source_shipping_amount"] = (
    order_header["shipping_amount_clean"]
)

order_header["source_discount_amount"] = (
    order_header["discount_amount_clean"]
)

order_header["source_tax_amount"] = (
    order_header["tax_amount_clean"]
)
```

```python
# Add source metadata and plausible missing values

order_header["source_file"] = (
    "orders_"
    + order_header["partner_id_clean"]
    + "_"
    + order_header["order_timestamp_clean"]
    .dt.strftime("%Y%m%d")
    + ".csv"
)

order_header["ingested_at"] = (
    order_header["order_timestamp_clean"]
    + pd.to_timedelta(
        order_rng.integers(
            1,
            48,
            size=N_ORDERS
        ),
        unit="h"
    )
)

# Some orders have no explicit shipping charge because the partner
# bundles shipping into the item amount.
shipping_missing_mask = order_rng.random(N_ORDERS) < 0.012

order_header.loc[
    shipping_missing_mask,
    "source_shipping_amount"
] = np.nan

# Some source files do not provide tax separately.
tax_missing_mask = order_rng.random(N_ORDERS) < 0.018

order_header.loc[
    tax_missing_mask,
    "source_tax_amount"
] = np.nan

# A small number of source statuses are blank and require investigation.
status_missing_mask = order_rng.random(N_ORDERS) < 0.006

order_header.loc[
    status_missing_mask,
    "source_status"
] = None

print("Source-style order-header fields created.")
display(order_header.head())
```

```python
# Keep only the external order-header fields

raw_order_header_export = order_header[
    [
        "source_order_id",
        "source_partner_code",
        "source_market",
        "source_created_at",
        "source_currency",
        "source_status",
        "source_item_amount",
        "source_shipping_amount",
        "source_discount_amount",
        "source_tax_amount",
        "source_file",
        "ingested_at"
    ]
].copy()

order_header_path = RAW_DIR / "order_header_export.csv"

raw_order_header_export.to_csv(
    order_header_path,
    index=False
)

print(f"Order-header export saved to: {order_header_path}")
print(f"Row count: {len(raw_order_header_export):,}")
print(
    f"File size in MB: "
    f"{order_header_path.stat().st_size / 1024**2:.2f}"
)
```

## Generate an unfamiliar order-line export

This file contains one row per product line within an order.

The source uses partner product codes rather than internal SKUs. We will map these codes back to the internal product catalogue during the cleaning and validation stage.

```python
# Prepare partner product choices for order-line generation

line_rng = np.random.default_rng(RANDOM_SEED + 60)

partner_mapping_for_lines = (
    partner_product_map
    .merge(
        product_ref[
            [
                "internal_sku",
                "unit_cost_gbp",
                "category"
            ]
        ],
        on="internal_sku",
        how="left"
    )
)

partner_sku_lookup = (
    partner_mapping_for_lines
    .groupby("partner_id")["partner_sku"]
    .apply(list)
    .to_dict()
)

partner_sku_to_internal = dict(
    zip(
        partner_mapping_for_lines["partner_sku"],
        partner_mapping_for_lines["internal_sku"]
    )
)

partner_sku_to_cost = dict(
    zip(
        partner_mapping_for_lines["partner_sku"],
        partner_mapping_for_lines["unit_cost_gbp"]
    )
)

print("Partner SKU lists prepared.")

for partner_id, sku_list in partner_sku_lookup.items():
    print(f"{partner_id}: {len(sku_list):,} available partner SKUs")
```

```python
# Assign one to four product lines to each order

line_count_per_order = line_rng.choice(
    [1, 2, 3, 4],
    size=len(order_header),
    p=[0.48, 0.34, 0.14, 0.04]
)

order_lines = (
    order_header
    .loc[
        order_header.index.repeat(line_count_per_order)
    ]
    .reset_index(drop=True)
)

order_lines["line_number"] = (
    order_lines
    .groupby("source_order_id")
    .cumcount()
    + 1
)

n_order_lines = len(order_lines)

print("Order headers:", f"{len(order_header):,}")
print("Order lines:", f"{n_order_lines:,}")
print(
    "Average lines per order:",
    round(n_order_lines / len(order_header), 2)
)
```

```python
# Select a partner SKU that is available for each order's partner

selected_partner_skus = [
    line_rng.choice(
        partner_sku_lookup[partner_id]
    )
    for partner_id in order_lines["partner_id_clean"]
]

order_lines["partner_sku_clean"] = selected_partner_skus

order_lines["internal_sku_clean"] = (
    order_lines["partner_sku_clean"]
    .map(partner_sku_to_internal)
)

order_lines["unit_cost_gbp"] = (
    order_lines["partner_sku_clean"]
    .map(partner_sku_to_cost)
)

# Most lines have a quantity of one, with a smaller number of multi-unit lines
order_lines["quantity_clean"] = line_rng.choice(
    [1, 2, 3, 4],
    size=n_order_lines,
    p=[0.70, 0.20, 0.08, 0.02]
)

# Create an initial commercial value for each line
line_markup = line_rng.uniform(
    1.8,
    3.3,
    size=n_order_lines
)

order_lines["initial_line_value"] = (
    order_lines["unit_cost_gbp"]
    * line_markup
    * order_lines["quantity_clean"]
)
```

```python
# Scale line values so their total approximately matches each order's item amount

order_lines["initial_order_value"] = (
    order_lines
    .groupby("source_order_id")["initial_line_value"]
    .transform("sum")
)

order_lines["line_value_clean"] = (
    order_lines["initial_line_value"]
    / order_lines["initial_order_value"]
    * order_lines["item_amount_clean"]
).round(2)

order_lines["unit_price_clean"] = (
    order_lines["line_value_clean"]
    / order_lines["quantity_clean"]
).round(2)

# Allocate the header discount across the order lines
order_lines["line_discount_clean"] = (
    order_lines["discount_amount_clean"]
    * order_lines["line_value_clean"]
    / order_lines
        .groupby("source_order_id")["line_value_clean"]
        .transform("sum")
).round(2)

print("Line-level commercial values created.")

display(
    order_lines[
        [
            "source_order_id",
            "line_number",
            "partner_sku_clean",
            "internal_sku_clean",
            "quantity_clean",
            "unit_price_clean",
            "line_discount_clean"
        ]
    ].head(10)
)
```

```python
# Convert clean order-line values into source-style fields

raw_order_lines = pd.DataFrame()

raw_order_lines["source_order_id"] = (
    order_lines["source_order_id"]
)

raw_order_lines["source_line_id"] = (
    order_lines["source_order_id"]
    + "-L"
    + order_lines["line_number"]
    .astype(str)
    .str.zfill(2)
)

raw_order_lines["source_partner_code"] = np.where(
    line_rng.random(n_order_lines) < 0.022,
    " " + order_lines["partner_id_clean"] + " ",
    order_lines["partner_id_clean"]
)

raw_order_lines["source_item_code"] = (
    order_lines["partner_sku_clean"]
)

item_code_whitespace_mask = line_rng.random(
    n_order_lines
) < 0.028

raw_order_lines.loc[
    item_code_whitespace_mask,
    "source_item_code"
] = (
    " "
    + raw_order_lines.loc[
        item_code_whitespace_mask,
        "source_item_code"
    ]
    + " "
)

raw_order_lines["source_quantity"] = (
    order_lines["quantity_clean"]
)

raw_order_lines["source_unit_price"] = (
    order_lines["unit_price_clean"]
)

raw_order_lines["source_line_discount"] = (
    order_lines["line_discount_clean"]
)

raw_order_lines["source_currency"] = (
    order_lines["source_currency"]
)

raw_order_lines["source_file"] = (
    "order_lines_"
    + order_lines["partner_id_clean"]
    + "_"
    + order_lines["order_timestamp_clean"]
    .dt.strftime("%Y%m%d")
    + ".csv"
)

raw_order_lines["ingested_at"] = (
    order_lines["ingested_at"]
    + pd.to_timedelta(
        line_rng.integers(
            0,
            12,
            size=n_order_lines
        ),
        unit="h"
    )
)
```

## Generate an unfamiliar shipment-event export

This file contains operational events for order lines. Each order line can have multiple events, such as accepted, shipped, delivered, or delivery exception.

The event data will later be used to calculate fulfilment and delivery metrics.

```python
# Prepare the base records for shipment-event generation

shipment_rng = np.random.default_rng(RANDOM_SEED + 70)

shipment_base = order_lines[
    [
        "source_order_id",
        "line_number",
        "partner_id_clean",
        "country_code_clean",
        "order_timestamp_clean",
        "order_status_clean"
    ]
].copy()

shipment_base["source_line_id"] = (
    shipment_base["source_order_id"]
    + "-L"
    + shipment_base["line_number"]
    .astype(str)
    .str.zfill(2)
)

shipment_base["order_age_days"] = (
    pd.Timestamp(PROJECT_END)
    - shipment_base["order_timestamp_clean"].dt.normalize()
).dt.days

n_shipment_lines = len(shipment_base)

print("Shipment line base rows:", f"{n_shipment_lines:,}")
display(shipment_base.head())
```

```python
# Create different shipment event types

shipment_events = []

# Every non-cancelled order line is normally accepted
accepted_lines = shipment_base[
    shipment_base["order_status_clean"] != "cancelled"
].copy()

accepted_lines["event_type_clean"] = "ACCEPTED"
accepted_lines["event_delay_days"] = shipment_rng.integers(
    0,
    2,
    size=len(accepted_lines)
)

shipment_events.append(accepted_lines)

# Most accepted lines are shipped
shippable_lines = accepted_lines[
    (
        shipment_rng.random(len(accepted_lines))
        < 0.90
    )
].copy()

shippable_lines["event_type_clean"] = "SHIPPED"
shippable_lines["event_delay_days"] = shipment_rng.integers(
    1,
    6,
    size=len(shippable_lines)
)

shipment_events.append(shippable_lines)

# Older shipped lines are more likely to have a delivery event
delivered_lines = shippable_lines[
    (
        shippable_lines["order_age_days"] >= 5
    )
    & (
        shipment_rng.random(len(shippable_lines))
        < 0.91
    )
].copy()

delivered_lines["event_type_clean"] = "DELIVERED"
delivered_lines["event_delay_days"] = shipment_rng.integers(
    3,
    12,
    size=len(delivered_lines)
)

shipment_events.append(delivered_lines)

# A smaller number of shipments receive a delivery exception
exception_lines = shippable_lines[
    shipment_rng.random(len(shippable_lines)) < 0.035
].copy()

exception_lines["event_type_clean"] = "DELIVERY_EXCEPTION"
exception_lines["event_delay_days"] = shipment_rng.integers(
    3,
    14,
    size=len(exception_lines)
)

shipment_events.append(exception_lines)

shipment_events = pd.concat(
    shipment_events,
    ignore_index=True
)

print("Shipment events created:", f"{len(shipment_events):,}")
print("\nEvent types:")
display(
    shipment_events["event_type_clean"]
    .value_counts()
    .to_frame("event_count")
)
```

```python
# Add event timestamps, carriers, fulfilment locations, and tracking references

n_events = len(shipment_events)

shipment_events["event_timestamp_clean"] = (
    shipment_events["order_timestamp_clean"]
    + pd.to_timedelta(
        shipment_events["event_delay_days"],
        unit="D"
    )
    + pd.to_timedelta(
        shipment_rng.integers(
            0,
            24,
            size=n_events
        ),
        unit="h"
    )
)

shipment_events["carrier_clean"] = shipment_rng.choice(
    [
        "CAR_A",
        "CAR_B",
        "CAR_C",
        "CAR_D"
    ],
    size=n_events,
    p=[0.35, 0.30, 0.20, 0.15]
)

shipment_events["fulfilment_country_clean"] = (
    shipment_events["country_code_clean"]
)

fulfilment_override_mask = shipment_rng.random(
    n_events
) < 0.08

shipment_events.loc[
    fulfilment_override_mask,
    "fulfilment_country_clean"
] = shipment_rng.choice(
    [
        "GB",
        "DE",
        "FR",
        "NL"
    ],
    size=fulfilment_override_mask.sum()
)

shipment_events["promised_delivery_date_clean"] = (
    shipment_events["order_timestamp_clean"].dt.normalize()
    + pd.to_timedelta(
        shipment_rng.integers(
            3,
            9,
            size=n_events
        ),
        unit="D"
    )
)

# Create tracking references using pandas string operations

tracking_numbers = pd.Series(
    shipment_rng.integers(
        10000000,
        99999999,
        size=n_events
    ).astype(str),
    index=shipment_events.index,
    dtype="string"
)

tracking_relevant_mask = (
    shipment_events["event_type_clean"]
    .isin(
        [
            "SHIPPED",
            "DELIVERED",
            "DELIVERY_EXCEPTION"
        ]
    )
)

tracking_references = (
    "TRK-"
    + tracking_numbers
)

shipment_events["tracking_reference_clean"] = (
    tracking_references
    .where(
        tracking_relevant_mask,
        pd.NA
    )
)

print("Operational shipment fields created.")
display(shipment_events.head())
```

```python
# Repair the missing tracking-reference column

tracking_numbers = pd.Series(
    shipment_rng.integers(
        10000000,
        99999999,
        size=len(shipment_events)
    ).astype(str),
    index=shipment_events.index,
    dtype="string"
)

tracking_relevant_mask = (
    shipment_events["event_type_clean"]
    .isin(
        [
            "SHIPPED",
            "DELIVERED",
            "DELIVERY_EXCEPTION"
        ]
    )
)

shipment_events["tracking_reference_clean"] = (
    (
        "TRK-"
        + tracking_numbers
    )
    .where(
        tracking_relevant_mask,
        pd.NA
    )
)

print("Tracking-reference column repaired.")
print(
    "Column exists:",
    "tracking_reference_clean" in shipment_events.columns
)
```

```python
# Convert shipment events into an unfamiliar source format

def format_event_timestamp(value, format_number):
    if format_number == 0:
        return value.strftime("%Y-%m-%d %H:%M:%S")
    elif format_number == 1:
        return value.strftime("%d/%m/%Y %H:%M")
    else:
        return value.strftime("%Y-%m-%dT%H:%M:%S")


shipment_export = pd.DataFrame()

shipment_export["external_order_id"] = (
    shipment_events["source_order_id"]
)

shipment_export["line_ref"] = (
    shipment_events["source_line_id"]
)

shipment_export["event"] = (
    shipment_events["event_type_clean"]
)

shipment_export["carrier"] = (
    shipment_events["carrier_clean"]
)

shipment_export["tracking_number"] = (
    shipment_events["tracking_reference_clean"]
)

shipment_export["fulfilment_market"] = (
    shipment_events["fulfilment_country_clean"]
    .map({
        "GB": "UK",
        "DE": "DE",
        "FR": "FR",
        "NL": "NLD",
        "IE": "IE",
        "ES": "ESP",
        "IT": "IT",
        "SE": "SWE",
        "DK": "DK",
        "BE": "BE"
    })
)

event_date_formats = shipment_rng.integers(
    0,
    3,
    size=n_events
)

shipment_export["event_time"] = [
    format_event_timestamp(
        timestamp,
        int(format_number)
    )
    for timestamp, format_number in zip(
        shipment_events["event_timestamp_clean"],
        event_date_formats
    )
]

promised_date_formats = shipment_rng.integers(
    0,
    3,
    size=n_events
)

shipment_export["promised_date"] = [
    format_event_timestamp(
        timestamp,
        int(format_number)
    )
    for timestamp, format_number in zip(
        shipment_events["promised_delivery_date_clean"],
        promised_date_formats
    )
]

shipment_export["source_file"] = (
    "shipment_events_"
    + shipment_events["event_timestamp_clean"]
    .dt.strftime("%Y%m%d")
    + ".csv"
)

shipment_export["ingested_at"] = (
    shipment_events["event_timestamp_clean"]
    + pd.to_timedelta(
        shipment_rng.integers(
            1,
            36,
            size=n_events
        ),
        unit="h"
    )
)
```

```python
# Add plausible missing values to shipment events

tracking_missing_mask = shipment_rng.random(
    n_events
) < 0.018

carrier_missing_mask = shipment_rng.random(
    n_events
) < 0.009

promised_date_missing_mask = shipment_rng.random(
    n_events
) < 0.012

shipment_export.loc[
    tracking_missing_mask,
    "tracking_number"
] = None

shipment_export.loc[
    carrier_missing_mask,
    "carrier"
] = None

shipment_export.loc[
    promised_date_missing_mask,
    "promised_date"
] = None

shipment_event_path = RAW_DIR / "shipment_event_export.csv"

shipment_export.to_csv(
    shipment_event_path,
    index=False
)

print(f"Shipment event export saved to: {shipment_event_path}")
print(f"Row count: {len(shipment_export):,}")
print(
    f"File size in MB: "
    f"{shipment_event_path.stat().st_size / 1024**2:.2f}"
)
```

```python
# Inspect the raw shipment-event export

print("Shipment-event columns:")
for column in shipment_export.columns:
    print(f" - {column}")

print("\nEvent counts:")
display(
    shipment_export["event"]
    .value_counts()
    .to_frame("event_count")
)

print("\nMissing values:")
display(
    shipment_export.isna()
    .sum()
    .to_frame("missing_count")
    .sort_values(
        "missing_count",
        ascending=False
    )
)

print("\nSample rows:")
print(shipment_export.head(10).to_string())
```

## Generate an unfamiliar returns export

This file represents return cases received from fictional partner channels.

Returns may be partial, delayed, associated with different reasons, or reflected in settlement data after the original order date.

```python
# Select synthetic return cases from the order-line population

return_rng = np.random.default_rng(RANDOM_SEED + 80)

return_base = order_lines[
    [
        "source_order_id",
        "line_number",
        "partner_id_clean",
        "country_code_clean",
        "order_timestamp_clean",
        "order_status_clean",
        "line_value_clean",
        "source_currency"
    ]
].copy()

return_base["source_line_id"] = (
    return_base["source_order_id"]
    + "-L"
    + return_base["line_number"]
    .astype(str)
    .str.zfill(2)
)

# Lines from returned or refunded orders are more likely to have a return case
forced_return_mask = return_base[
    "order_status_clean"
].isin(
    [
        "returned",
        "refunded"
    ]
)

additional_return_mask = (
    return_base["order_status_clean"]
    .isin(
        [
            "delivered",
            "shipped"
        ]
    )
    & (
        return_rng.random(len(return_base))
        < 0.105
    )
)

return_case_base = return_base[
    forced_return_mask | additional_return_mask
].copy()

return_case_base = return_case_base.reset_index(
    drop=True
)

n_returns = len(return_case_base)

print("Return cases selected:", f"{n_returns:,}")
display(return_case_base.head())
```

```python
# Create synthetic return details

return_case_base["return_reason_clean"] = return_rng.choice(
    [
        "size_or_fit",
        "changed_mind",
        "not_as_described",
        "damaged_in_transit",
        "quality_issue",
        "wrong_item",
        "late_delivery"
    ],
    size=n_returns,
    p=[
        0.30,
        0.21,
        0.13,
        0.10,
        0.12,
        0.08,
        0.06
    ]
)

return_case_base["return_request_date_clean"] = (
    return_case_base["order_timestamp_clean"]
    + pd.to_timedelta(
        return_rng.integers(
            5,
            45,
            size=n_returns
        ),
        unit="D"
    )
)

return_case_base["return_received_date_clean"] = (
    return_case_base["return_request_date_clean"]
    + pd.to_timedelta(
        return_rng.integers(
            2,
            18,
            size=n_returns
        ),
        unit="D"
    )
)

return_case_base["refund_amount_clean"] = (
    return_case_base["line_value_clean"]
    * return_rng.uniform(
        0.75,
        1.00,
        size=n_returns
    )
).round(2)

return_case_base["return_disposition_clean"] = return_rng.choice(
    [
        "resalable",
        "discounted_resale",
        "damaged",
        "supplier_review",
        "written_off"
    ],
    size=n_returns,
    p=[
        0.57,
        0.18,
        0.10,
        0.08,
        0.07
    ]
)

return_case_base["return_status_clean"] = return_rng.choice(
    [
        "requested",
        "in_transit",
        "received",
        "refunded",
        "rejected"
    ],
    size=n_returns,
    p=[
        0.08,
        0.10,
        0.14,
        0.64,
        0.04
    ]
)

print("Return details created.")
display(
    return_case_base[
        [
            "return_reason_clean",
            "return_disposition_clean",
            "return_status_clean"
        ]
    ]
    .value_counts()
    .head(15)
    .to_frame("case_count")
)
```

```python
# Convert returns into an unfamiliar source format

def format_return_date(value, format_number):
    if format_number == 0:
        return value.strftime("%Y-%m-%d")
    elif format_number == 1:
        return value.strftime("%d/%m/%Y")
    else:
        return value.strftime("%Y-%m-%d %H:%M:%S")


return_export = pd.DataFrame()

return_export["return_id"] = [
    f"RET-{number:09d}"
    for number in range(1, n_returns + 1)
]

return_export["external_order_id"] = (
    return_case_base["source_order_id"]
)

return_export["line_ref"] = (
    return_case_base["source_line_id"]
)

return_export["partner_code"] = np.where(
    return_rng.random(n_returns) < 0.022,
    " " + return_case_base["partner_id_clean"] + " ",
    return_case_base["partner_id_clean"]
)

return_export["market"] = (
    return_case_base["country_code_clean"]
    .map({
        "GB": "UK",
        "IE": "IE",
        "DE": "DE",
        "FR": "FR",
        "NL": "NLD",
        "BE": "BE",
        "ES": "ESP",
        "IT": "IT",
        "SE": "SWE",
        "DK": "DK"
    })
)

return_export["reason"] = (
    return_case_base["return_reason_clean"]
)

return_export["status"] = (
    return_case_base["return_status_clean"]
)

return_export["disposition"] = (
    return_case_base["return_disposition_clean"]
)

request_date_formats = return_rng.integers(
    0,
    3,
    size=n_returns
)

return_export["requested_at"] = [
    format_return_date(
        return_date,
        int(format_number)
    )
    for return_date, format_number in zip(
        return_case_base["return_request_date_clean"],
        request_date_formats
    )
]

received_date_formats = return_rng.integers(
    0,
    3,
    size=n_returns
)

return_export["received_at"] = [
    format_return_date(
        return_date,
        int(format_number)
    )
    for return_date, format_number in zip(
        return_case_base["return_received_date_clean"],
        received_date_formats
    )
]

return_export["refund_amount"] = (
    return_case_base["refund_amount_clean"]
)

return_export["currency"] = (
    return_case_base["source_currency"]
)

return_export["source_file"] = (
    "returns_"
    + return_case_base["partner_id_clean"]
    + "_"
    + return_case_base["return_request_date_clean"]
    .dt.strftime("%Y%m%d")
    + ".csv"
)

return_export["ingested_at"] = (
    return_case_base["return_received_date_clean"]
    + pd.to_timedelta(
        return_rng.integers(
            1,
            72,
            size=n_returns
        ),
        unit="h"
    )
)
```

## Generate synthetic FX-rate data

This file contains synthetic daily exchange rates to GBP.

The rates are created only for analytical demonstration. They are not historical market rates and should not be treated as financial or tax advice.

```python
# Create synthetic daily FX rates to GBP

fx_rng = np.random.default_rng(RANDOM_SEED + 100)

fx_dates = pd.date_range(
    start=PROJECT_START,
    end=PROJECT_END,
    freq="D"
)

fx_currency_parameters = {
    "GBP": {
        "starting_rate": 1.0000,
        "volatility": 0.0000
    },
    "EUR": {
        "starting_rate": 0.8600,
        "volatility": 0.0025
    },
    "SEK": {
        "starting_rate": 0.0740,
        "volatility": 0.0040
    },
    "DKK": {
        "starting_rate": 0.1150,
        "volatility": 0.0015
    }
}

fx_rows = []

for currency_code, parameters in fx_currency_parameters.items():
    current_rate = parameters["starting_rate"]

    for rate_date in fx_dates:
        daily_change = fx_rng.normal(
            loc=0.0,
            scale=parameters["volatility"]
        )

        current_rate = max(
            current_rate * (1 + daily_change),
            parameters["starting_rate"] * 0.75
        )

        fx_rows.append({
            "rate_date_clean": rate_date,
            "from_currency_clean": currency_code,
            "to_currency_clean": "GBP",
            "rate_to_gbp_clean": round(
                float(current_rate),
                6
            ),
            "rate_source_clean": "synthetic_daily_rate"
        })

fx_clean = pd.DataFrame(fx_rows)

print("FX rows created:", f"{len(fx_clean):,}")
display(fx_clean.head())
```

```python
# Convert FX data into an unfamiliar source format

def format_fx_date(value, format_number):
    if format_number == 0:
        return value.strftime("%Y-%m-%d")
    elif format_number == 1:
        return value.strftime("%d/%m/%Y")
    else:
        return value.strftime("%Y-%m-%d %H:%M:%S")


n_fx_rows = len(fx_clean)

fx_export = pd.DataFrame()

fx_export["rate_date"] = [
    format_fx_date(
        date_value,
        int(format_number)
    )
    for date_value, format_number in zip(
        fx_clean["rate_date_clean"],
        fx_rng.integers(
            0,
            3,
            size=n_fx_rows
        )
    )
]

fx_export["source_currency"] = (
    fx_clean["from_currency_clean"]
)

fx_export["reporting_currency"] = (
    fx_clean["to_currency_clean"]
)

fx_export["conversion_rate"] = (
    fx_clean["rate_to_gbp_clean"]
)

fx_export["rate_source"] = (
    fx_clean["rate_source_clean"]
)

fx_export["source_file"] = (
    "fx_rates_"
    + fx_clean["rate_date_clean"]
    .dt.strftime("%Y%m%d")
    + ".csv"
)

fx_export["ingested_at"] = (
    fx_clean["rate_date_clean"]
    + pd.to_timedelta(
        fx_rng.integers(
            1,
            24,
            size=n_fx_rows
        ),
        unit="h"
    )
)

print("Source-style FX fields created.")
display(fx_export.head())
```

## Build Silver listing and inventory tables

The Silver layer standardises raw values while preserving source lineage and data-quality outcomes.

Records are not automatically deleted. Each record receives a status such as valid, review, or quarantine.

```python
# Helper functions for standardising raw source values

def parse_mixed_datetime(series):
    return pd.to_datetime(
        series.astype("string").str.strip(),
        errors="coerce",
        format="mixed",
        dayfirst=True
    )


def parse_mixed_number(series):
    return pd.to_numeric(
        series
        .astype("string")
        .str.strip()
        .str.replace("£", "", regex=False)
        .str.replace("€", "", regex=False)
        .str.replace(" ", "", regex=False)
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )


country_alias_map = {
    "GB": "GB",
    "UK": "GB",
    "GBR": "GB",
    "IE": "IE",
    "DE": "DE",
    "FR": "FR",
    "NL": "NL",
    "NLD": "NL",
    "BE": "BE",
    "ES": "ES",
    "ESP": "ES",
    "IT": "IT",
    "SE": "SE",
    "SWE": "SE",
    "DK": "DK"
}

partner_code_map = {
    partner_id: partner_id
    for partner_id in partner_ref["partner_id"]
}

canonical_status_map = {
    "LIVE": "LIVE",
    "OFFLINE": "OFFLINE",
    "PAUSED": "PAUSED",
    "PREORDER": "PREORDER"
}

country_currency_map = (
    country_ref
    .set_index("country_code")["currency_code"]
    .to_dict()
)


def standardise_currency(currency_value, country_code):
    if pd.isna(currency_value):
        return pd.NA

    value = str(currency_value).strip()

    if value in ["GBP", "£", "Pound Sterling"]:
        return "GBP"

    if value in ["EUR", "€", "Euro"]:
        return "EUR"

    if value in ["SEK", "Swedish Krona"]:
        return "SEK"

    if value in ["DKK", "Danish Krone"]:
        return "DKK"

    # The alias "kr" is ambiguous, so use the country context.
    if value == "kr":
        return country_currency_map.get(
            country_code,
            pd.NA
        )

    return pd.NA


def combine_reasons(reason_series):
    reasons = [
        str(value)
        for value in reason_series
        if pd.notna(value) and str(value) != ""
    ]

    if not reasons:
        return pd.NA

    return "; ".join(reasons)
```

```python
# Load and standardise the raw listing export

raw_listing = raw_tables[
    "partner_listing_export"
].copy()

silver_listing = pd.DataFrame()

silver_listing["source_row_number"] = (
    raw_listing.index + 1
)

silver_listing["source_partner_code"] = (
    raw_listing["source_partner_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_listing["partner_id"] = (
    silver_listing["source_partner_code"]
    .map(partner_code_map)
)

silver_listing["source_item_code"] = (
    raw_listing["source_item_code"]
    .astype("string")
    .str.strip()
)

silver_listing["source_market"] = (
    raw_listing["source_market"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_listing["country_code"] = (
    silver_listing["source_market"]
    .map(country_alias_map)
)

silver_listing["snapshot_date"] = (
    parse_mixed_datetime(
        raw_listing["source_snapshot_date"]
    )
    .dt.normalize()
)

silver_listing["listing_status"] = (
    raw_listing["source_listing_state"]
    .astype("string")
    .str.strip()
    .str.upper()
    .map(canonical_status_map)
)

silver_listing["price_local"] = (
    parse_mixed_number(
        raw_listing["source_price"]
    )
)

silver_listing["currency_code"] = [
    standardise_currency(
        currency_value,
        country_code
    )
    for currency_value, country_code in zip(
        raw_listing["source_currency"],
        silver_listing["country_code"]
    )
]

silver_listing["available_units"] = (
    parse_mixed_number(
        raw_listing["source_stock"]
    )
)

silver_listing["inventory_updated_at"] = (
    parse_mixed_datetime(
        raw_listing["source_inventory_update"]
    )
)

silver_listing["title"] = (
    raw_listing["source_title"]
)

silver_listing["category_source"] = (
    raw_listing["source_category"]
    .astype("string")
    .str.strip()
)

silver_listing["image_url"] = (
    raw_listing["source_image_url"]
)

silver_listing["source_file"] = (
    raw_listing["source_file"]
)

silver_listing["ingested_at"] = (
    parse_mixed_datetime(
        raw_listing["ingested_at"]
    )
)

print("Listing source values standardised.")
display(silver_listing.head())
```

```python
# Map partner item codes to internal products

mapping_lookup = (
    partner_product_map[
        [
            "partner_id",
            "partner_sku",
            "internal_sku"
        ]
    ]
    .drop_duplicates()
)

silver_listing = silver_listing.merge(
    mapping_lookup.rename(
        columns={
            "partner_sku": "source_item_code"
        }
    ),
    on=[
        "partner_id",
        "source_item_code"
    ],
    how="left"
)

# Create product content completeness
silver_listing["title_present"] = (
    silver_listing["title"]
    .notna()
    & silver_listing["title"]
    .astype("string")
    .str.strip()
    .ne("")
)

silver_listing["category_present"] = (
    silver_listing["category_source"]
    .notna()
    & silver_listing["category_source"]
    .astype("string")
    .str.strip()
    .ne("")
)

silver_listing["image_present"] = (
    silver_listing["image_url"]
    .notna()
    & silver_listing["image_url"]
    .astype("string")
    .str.strip()
    .ne("")
)

silver_listing["content_completeness_pct"] = (
    silver_listing[
        [
            "title_present",
            "category_present",
            "image_present"
        ]
    ]
    .mean(axis=1)
    * 100
).round(2)

# Determine whether the record is a possible repeated business key
silver_listing["listing_business_key"] = (
    silver_listing["partner_id"].astype("string")
    + "|"
    + silver_listing["source_item_code"].astype("string")
    + "|"
    + silver_listing["country_code"].astype("string")
    + "|"
    + silver_listing["snapshot_date"].astype("string")
)

silver_listing["possible_duplicate_key"] = (
    silver_listing["listing_business_key"]
    .duplicated(keep=False)
)

print("Listing mappings and quality flags created.")
```

```python
# Apply listing data-quality rules

silver_listing["quality_reasons"] = pd.NA

listing_reason_rows = []

for _, row in silver_listing.iterrows():
    reasons = []

    if pd.isna(row["partner_id"]):
        reasons.append("unrecognised_partner")

    if (
        pd.isna(row["source_item_code"])
        or str(row["source_item_code"]).strip() == ""
    ):
        reasons.append("missing_item_code")

    if pd.isna(row["snapshot_date"]):
        reasons.append("invalid_snapshot_date")

    if pd.isna(row["country_code"]):
        reasons.append("unrecognised_country")

    if pd.isna(row["internal_sku"]):
        reasons.append("unmapped_partner_sku")

    if (
        pd.isna(row["price_local"])
        or row["price_local"] < 0
    ):
        reasons.append("invalid_price")

    if (
        pd.notna(row["available_units"])
        and row["available_units"] < 0
    ):
        reasons.append("negative_stock")

    if pd.isna(row["listing_status"]):
        reasons.append("unrecognised_listing_status")

    if row["possible_duplicate_key"]:
        reasons.append("possible_duplicate_business_key")

    listing_reason_rows.append(
        "; ".join(reasons)
        if reasons
        else pd.NA
    )

silver_listing["quality_reasons"] = (
    listing_reason_rows
)

critical_listing_reasons = [
    "unrecognised_partner",
    "missing_item_code",
    "invalid_snapshot_date",
    "unrecognised_country",
    "unmapped_partner_sku",
    "invalid_price"
]

review_listing_reasons = [
    "negative_stock",
    "unrecognised_listing_status",
    "possible_duplicate_business_key"
]

silver_listing["record_status"] = "valid"

for reason in critical_listing_reasons:
    silver_listing.loc[
        silver_listing["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False),
        "record_status"
    ] = "quarantine"

for reason in review_listing_reasons:
    review_mask = (
        silver_listing["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False)
        & (
            silver_listing["record_status"]
            == "valid"
        )
    )

    silver_listing.loc[
        review_mask,
        "record_status"
    ] = "review"

silver_listing["analysis_included"] = (
    silver_listing["record_status"]
    != "quarantine"
)

silver_listing["live_listing_flag"] = (
    silver_listing["listing_status"].eq("LIVE")
    & silver_listing["available_units"].gt(0)
    & silver_listing["analysis_included"]
)

print("Listing records by status:")
display(
    silver_listing["record_status"]
    .value_counts()
    .to_frame("record_count")
)
```

```python
# Separate listing records requiring quarantine

quarantine_listing = silver_listing[
    silver_listing["record_status"] == "quarantine"
].copy()

print(
    "Listing quarantine records:",
    f"{len(quarantine_listing):,}"
)

display(
    quarantine_listing[
        [
            "source_row_number",
            "source_partner_code",
            "source_item_code",
            "country_code",
            "internal_sku",
            "quality_reasons"
        ]
    ]
    .head(10)
)
```

```python
# Load and standardise the raw inventory export

raw_inventory = raw_tables[
    "inventory_snapshot_export"
].copy()

silver_inventory = pd.DataFrame()

silver_inventory["source_row_number"] = (
    raw_inventory.index + 1
)

silver_inventory["source_partner_code"] = (
    raw_inventory["partner_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_inventory["partner_id"] = (
    silver_inventory["source_partner_code"]
    .map(partner_code_map)
)

silver_inventory["source_item_code"] = (
    raw_inventory["item_code"]
    .astype("string")
    .str.strip()
)

silver_inventory["source_market"] = (
    raw_inventory["market"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_inventory["country_code"] = (
    silver_inventory["source_market"]
    .map(country_alias_map)
)

silver_inventory["warehouse_code"] = (
    raw_inventory["warehouse"]
    .astype("string")
    .str.strip()
)

silver_inventory["available_units"] = (
    parse_mixed_number(
        raw_inventory["available_qty"]
    )
)

silver_inventory["reserved_units"] = (
    parse_mixed_number(
        raw_inventory["reserved_qty"]
    )
)

silver_inventory["inbound_units"] = (
    parse_mixed_number(
        raw_inventory["inbound_qty"]
    )
)

silver_inventory["inventory_timestamp"] = (
    parse_mixed_datetime(
        raw_inventory["inventory_timestamp"]
    )
)

silver_inventory["source_file"] = (
    raw_inventory["source_file"]
)

silver_inventory["ingested_at"] = (
    parse_mixed_datetime(
        raw_inventory["ingested_at"]
    )
)

silver_inventory = silver_inventory.merge(
    mapping_lookup.rename(
        columns={
            "partner_sku": "source_item_code"
        }
    ),
    on=[
        "partner_id",
        "source_item_code"
    ],
    how="left"
)

print("Inventory source values standardised.")
display(silver_inventory.head())
```

```python
# Apply inventory data-quality rules

silver_inventory["inventory_business_key"] = (
    silver_inventory["partner_id"].astype("string")
    + "|"
    + silver_inventory["source_item_code"].astype("string")
    + "|"
    + silver_inventory["country_code"].astype("string")
    + "|"
    + silver_inventory["warehouse_code"].astype("string")
    + "|"
    + silver_inventory["inventory_timestamp"].astype("string")
)

silver_inventory["possible_duplicate_key"] = (
    silver_inventory["inventory_business_key"]
    .duplicated(keep=False)
)

inventory_reason_rows = []

for _, row in silver_inventory.iterrows():
    reasons = []

    if pd.isna(row["partner_id"]):
        reasons.append("unrecognised_partner")

    if (
        pd.isna(row["source_item_code"])
        or str(row["source_item_code"]).strip() == ""
    ):
        reasons.append("missing_item_code")

    if pd.isna(row["country_code"]):
        reasons.append("unrecognised_country")

    if pd.isna(row["internal_sku"]):
        reasons.append("unmapped_partner_sku")

    if pd.isna(row["inventory_timestamp"]):
        reasons.append("invalid_inventory_timestamp")

    for quantity_column in [
        "available_units",
        "reserved_units",
        "inbound_units"
    ]:
        if (
            pd.notna(row[quantity_column])
            and row[quantity_column] < 0
        ):
            reasons.append(
                f"negative_{quantity_column}"
            )

    if row["possible_duplicate_key"]:
        reasons.append("possible_duplicate_business_key")

    inventory_reason_rows.append(
        "; ".join(reasons)
        if reasons
        else pd.NA
    )

silver_inventory["quality_reasons"] = (
    inventory_reason_rows
)

critical_inventory_reasons = [
    "unrecognised_partner",
    "missing_item_code",
    "unrecognised_country",
    "unmapped_partner_sku",
    "invalid_inventory_timestamp"
]

review_inventory_reasons = [
    "negative_available_units",
    "negative_reserved_units",
    "negative_inbound_units",
    "possible_duplicate_business_key"
]

silver_inventory["record_status"] = "valid"

for reason in critical_inventory_reasons:
    silver_inventory.loc[
        silver_inventory["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False),
        "record_status"
    ] = "quarantine"

for reason in review_inventory_reasons:
    review_mask = (
        silver_inventory["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False)
        & (
            silver_inventory["record_status"]
            == "valid"
        )
    )

    silver_inventory.loc[
        review_mask,
        "record_status"
    ] = "review"

silver_inventory["analysis_included"] = (
    silver_inventory["record_status"]
    != "quarantine"
)

print("Inventory records by status:")
display(
    silver_inventory["record_status"]
    .value_counts()
    .to_frame("record_count")
)
```

## Build Silver order and order-line tables

Order headers and order lines are standardised separately, then checked against each other.

The final reconciliation compares the order-header merchandise value with the sum of the related order-line values.

```python
# Load and standardise the raw order-header export

raw_order_headers = raw_tables[
    "order_header_export"
].copy()

order_status_map = {
    "COMPLETED": "delivered",
    "DISPATCHED": "shipped",
    "OPEN": "pending",
    "CANCELLED": "cancelled",
    "RETURNED": "returned",
    "REFUNDED": "refunded"
}

silver_order_header = pd.DataFrame()

silver_order_header["source_row_number"] = (
    raw_order_headers.index + 1
)

silver_order_header["order_id"] = (
    raw_order_headers["source_order_id"]
    .astype("string")
    .str.strip()
)

silver_order_header["source_partner_code"] = (
    raw_order_headers["source_partner_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_order_header["partner_id"] = (
    silver_order_header["source_partner_code"]
    .map(partner_code_map)
)

silver_order_header["source_market"] = (
    raw_order_headers["source_market"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_order_header["country_code"] = (
    silver_order_header["source_market"]
    .map(country_alias_map)
)

silver_order_header["order_timestamp"] = (
    parse_mixed_datetime(
        raw_order_headers["source_created_at"]
    )
)

silver_order_header["order_status"] = (
    raw_order_headers["source_status"]
    .astype("string")
    .str.strip()
    .str.upper()
    .map(order_status_map)
)

silver_order_header["currency_code"] = [
    standardise_currency(
        currency_value,
        country_code
    )
    for currency_value, country_code in zip(
        raw_order_headers["source_currency"],
        silver_order_header["country_code"]
    )
]

silver_order_header["item_amount"] = (
    parse_mixed_number(
        raw_order_headers["source_item_amount"]
    )
)

silver_order_header["shipping_amount"] = (
    parse_mixed_number(
        raw_order_headers["source_shipping_amount"]
    )
)

silver_order_header["discount_amount"] = (
    parse_mixed_number(
        raw_order_headers["source_discount_amount"]
    )
)

silver_order_header["tax_amount"] = (
    parse_mixed_number(
        raw_order_headers["source_tax_amount"]
    )
)

silver_order_header["source_file"] = (
    raw_order_headers["source_file"]
)

silver_order_header["ingested_at"] = (
    parse_mixed_datetime(
        raw_order_headers["ingested_at"]
    )
)

silver_order_header["order_gross_amount"] = (
    silver_order_header["item_amount"]
    - silver_order_header["discount_amount"].fillna(0)
    + silver_order_header["shipping_amount"].fillna(0)
).round(2)

print("Order headers standardised.")
display(silver_order_header.head())
```

```python
# Apply data-quality rules to order headers

silver_order_header["possible_duplicate_order"] = (
    silver_order_header["order_id"]
    .duplicated(keep=False)
)

order_header_reason_rows = []

for _, row in silver_order_header.iterrows():
    reasons = []

    if (
        pd.isna(row["order_id"])
        or str(row["order_id"]).strip() == ""
    ):
        reasons.append("missing_order_id")

    if pd.isna(row["partner_id"]):
        reasons.append("unrecognised_partner")

    if pd.isna(row["country_code"]):
        reasons.append("unrecognised_country")

    if pd.isna(row["order_timestamp"]):
        reasons.append("invalid_order_timestamp")

    if pd.isna(row["order_status"]):
        reasons.append("unrecognised_order_status")

    if (
        pd.isna(row["item_amount"])
        or row["item_amount"] < 0
    ):
        reasons.append("invalid_item_amount")

    if (
        pd.notna(row["shipping_amount"])
        and row["shipping_amount"] < 0
    ):
        reasons.append("negative_shipping_amount")

    if (
        pd.notna(row["discount_amount"])
        and row["discount_amount"] < 0
    ):
        reasons.append("negative_discount_amount")

    if (
        pd.notna(row["tax_amount"])
        and row["tax_amount"] < 0
    ):
        reasons.append("negative_tax_amount")

    if row["possible_duplicate_order"]:
        reasons.append("possible_duplicate_order")

    order_header_reason_rows.append(
        "; ".join(reasons)
        if reasons
        else pd.NA
    )

silver_order_header["quality_reasons"] = (
    order_header_reason_rows
)

critical_order_header_reasons = [
    "missing_order_id",
    "unrecognised_partner",
    "unrecognised_country",
    "invalid_order_timestamp",
    "unrecognised_order_status",
    "invalid_item_amount"
]

review_order_header_reasons = [
    "negative_shipping_amount",
    "negative_discount_amount",
    "negative_tax_amount",
    "possible_duplicate_order"
]

silver_order_header["record_status"] = "valid"

for reason in critical_order_header_reasons:
    silver_order_header.loc[
        silver_order_header["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False),
        "record_status"
    ] = "quarantine"

for reason in review_order_header_reasons:
    review_mask = (
        silver_order_header["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False)
        & (
            silver_order_header["record_status"]
            == "valid"
        )
    )

    silver_order_header.loc[
        review_mask,
        "record_status"
    ] = "review"

silver_order_header["analysis_included"] = (
    silver_order_header["record_status"]
    != "quarantine"
)

print("Order-header records by status:")
display(
    silver_order_header["record_status"]
    .value_counts()
    .to_frame("record_count")
)
```

```python
# Load and standardise the raw order-line export

raw_order_lines = raw_tables[
    "order_line_export"
].copy()

silver_order_line = pd.DataFrame()

silver_order_line["source_row_number"] = (
    raw_order_lines.index + 1
)

silver_order_line["order_id"] = (
    raw_order_lines["source_order_id"]
    .astype("string")
    .str.strip()
)

silver_order_line["line_id"] = (
    raw_order_lines["source_line_id"]
    .astype("string")
    .str.strip()
)

silver_order_line["source_partner_code"] = (
    raw_order_lines["source_partner_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_order_line["partner_id"] = (
    silver_order_line["source_partner_code"]
    .map(partner_code_map)
)

silver_order_line["source_item_code"] = (
    raw_order_lines["source_item_code"]
    .astype("string")
    .str.strip()
)

silver_order_line["quantity"] = (
    parse_mixed_number(
        raw_order_lines["source_quantity"]
    )
)

silver_order_line["unit_price"] = (
    parse_mixed_number(
        raw_order_lines["source_unit_price"]
    )
)

silver_order_line["line_discount"] = (
    parse_mixed_number(
        raw_order_lines["source_line_discount"]
    )
    .fillna(0)
)

silver_order_line["currency_code"] = (
    raw_order_lines["source_currency"]
    .astype("string")
    .str.strip()
    .map({
        "GBP": "GBP",
        "£": "GBP",
        "Pound Sterling": "GBP",
        "EUR": "EUR",
        "€": "EUR",
        "Euro": "EUR",
        "SEK": "SEK",
        "kr": "SEK",
        "Swedish Krona": "SEK",
        "DKK": "DKK",
        "Danish Krone": "DKK"
    })
)

silver_order_line["source_file"] = (
    raw_order_lines["source_file"]
)

silver_order_line["ingested_at"] = (
    parse_mixed_datetime(
        raw_order_lines["ingested_at"]
    )
)

# Recalculate the line's gross and net merchandise amounts
silver_order_line["line_gross_amount"] = (
    silver_order_line["quantity"]
    * silver_order_line["unit_price"]
).round(2)

silver_order_line["line_net_amount"] = (
    silver_order_line["line_gross_amount"]
    - silver_order_line["line_discount"]
).round(2)

# Map partner item codes to internal products
silver_order_line = silver_order_line.merge(
    mapping_lookup.rename(
        columns={
            "partner_sku": "source_item_code"
        }
    ),
    on=[
        "partner_id",
        "source_item_code"
    ],
    how="left"
)

print("Order lines standardised.")
display(silver_order_line.head())
```

```python
# Apply data-quality rules to order lines

silver_order_line["possible_duplicate_line"] = (
    silver_order_line["line_id"]
    .duplicated(keep=False)
)

order_line_reason_rows = []

for _, row in silver_order_line.iterrows():
    reasons = []

    if (
        pd.isna(row["order_id"])
        or str(row["order_id"]).strip() == ""
    ):
        reasons.append("missing_order_id")

    if (
        pd.isna(row["line_id"])
        or str(row["line_id"]).strip() == ""
    ):
        reasons.append("missing_line_id")

    if pd.isna(row["partner_id"]):
        reasons.append("unrecognised_partner")

    if (
        pd.isna(row["source_item_code"])
        or str(row["source_item_code"]).strip() == ""
    ):
        reasons.append("missing_item_code")

    if pd.isna(row["internal_sku"]):
        reasons.append("unmapped_partner_sku")

    if (
        pd.isna(row["quantity"])
        or row["quantity"] <= 0
    ):
        reasons.append("invalid_quantity")

    if (
        pd.isna(row["unit_price"])
        or row["unit_price"] < 0
    ):
        reasons.append("invalid_unit_price")

    if (
        pd.notna(row["line_discount"])
        and row["line_discount"] < 0
    ):
        reasons.append("negative_line_discount")

    if row["possible_duplicate_line"]:
        reasons.append("possible_duplicate_line")

    order_line_reason_rows.append(
        "; ".join(reasons)
        if reasons
        else pd.NA
    )

silver_order_line["quality_reasons"] = (
    order_line_reason_rows
)

critical_order_line_reasons = [
    "missing_order_id",
    "missing_line_id",
    "unrecognised_partner",
    "missing_item_code",
    "unmapped_partner_sku",
    "invalid_quantity",
    "invalid_unit_price"
]

review_order_line_reasons = [
    "negative_line_discount",
    "possible_duplicate_line"
]

silver_order_line["record_status"] = "valid"

for reason in critical_order_line_reasons:
    silver_order_line.loc[
        silver_order_line["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False),
        "record_status"
    ] = "quarantine"

for reason in review_order_line_reasons:
    review_mask = (
        silver_order_line["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False)
        & (
            silver_order_line["record_status"]
            == "valid"
        )
    )

    silver_order_line.loc[
        review_mask,
        "record_status"
    ] = "review"

silver_order_line["analysis_included"] = (
    silver_order_line["record_status"]
    != "quarantine"
)

print("Order-line records by status:")
display(
    silver_order_line["record_status"]
    .value_counts()
    .to_frame("record_count")
)
```

```python
# Validate the relationship between order headers and order lines

valid_order_ids = set(
    silver_order_header.loc[
        silver_order_header["analysis_included"],
        "order_id"
    ]
    .dropna()
)

silver_order_line["orphan_order_flag"] = (
    ~silver_order_line["order_id"]
    .isin(valid_order_ids)
)

silver_order_line.loc[
    silver_order_line["orphan_order_flag"],
    "record_status"
] = "quarantine"

silver_order_line.loc[
    silver_order_line["orphan_order_flag"],
    "analysis_included"
] = False

orders_with_lines = set(
    silver_order_line.loc[
        silver_order_line["analysis_included"],
        "order_id"
    ]
    .dropna()
)

silver_order_header["missing_line_flag"] = (
    silver_order_header["analysis_included"]
    & ~silver_order_header["order_id"]
    .isin(orders_with_lines)
)

print(
    "Orphan order lines:",
    int(silver_order_line["orphan_order_flag"].sum())
)

print(
    "Orders without valid lines:",
    int(silver_order_header["missing_line_flag"].sum())
)
```

```python
# Reconcile header merchandise amounts to line-level amounts

valid_lines = silver_order_line[
    silver_order_line["analysis_included"]
].copy()

line_totals = (
    valid_lines
    .groupby("order_id", as_index=False)
    .agg(
        line_gross_amount=("line_gross_amount", "sum"),
        line_discount=("line_discount", "sum"),
        line_net_amount=("line_net_amount", "sum"),
        line_count=("line_id", "nunique")
    )
)

order_reconciliation = (
    silver_order_header[
        silver_order_header["analysis_included"]
    ][
        [
            "order_id",
            "item_amount",
            "discount_amount",
            "order_status",
            "partner_id",
            "country_code"
        ]
    ]
    .merge(
        line_totals,
        on="order_id",
        how="left"
    )
)

order_reconciliation["expected_net_merchandise"] = (
    order_reconciliation["item_amount"]
    - order_reconciliation["discount_amount"].fillna(0)
).round(2)

order_reconciliation["line_net_variance"] = (
    order_reconciliation["line_net_amount"]
    - order_reconciliation["expected_net_merchandise"]
).round(2)

order_reconciliation["reconciliation_status"] = np.select(
    [
        order_reconciliation["line_count"].isna(),
        order_reconciliation["line_net_variance"].abs() <= 0.05,
        order_reconciliation["line_net_variance"].abs() <= 1.00
    ],
    [
        "missing_lines",
        "within_tolerance",
        "minor_variance"
    ],
    default="investigate"
)

print("Order reconciliation statuses:")
display(
    order_reconciliation[
        "reconciliation_status"
    ]
    .value_counts()
    .to_frame("order_count")
)
```

## Build Silver shipment and return tables

Shipment events and returns are standardised and checked against valid order-line identifiers.

```python
# Load and standardise shipment events

raw_shipments = raw_tables[
    "shipment_event_export"
].copy()

shipment_event_map = {
    "ACCEPTED": "accepted",
    "SHIPPED": "shipped",
    "DELIVERED": "delivered",
    "DELIVERY_EXCEPTION": "delivery_exception"
}

silver_shipment_event = pd.DataFrame()

silver_shipment_event["source_row_number"] = (
    raw_shipments.index + 1
)

silver_shipment_event["order_id"] = (
    raw_shipments["external_order_id"]
    .astype("string")
    .str.strip()
)

silver_shipment_event["line_id"] = (
    raw_shipments["line_ref"]
    .astype("string")
    .str.strip()
)

silver_shipment_event["event_type"] = (
    raw_shipments["event"]
    .astype("string")
    .str.strip()
    .str.upper()
    .map(shipment_event_map)
)

silver_shipment_event["carrier_code"] = (
    raw_shipments["carrier"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_shipment_event["tracking_reference"] = (
    raw_shipments["tracking_number"]
    .astype("string")
    .str.strip()
)

silver_shipment_event["fulfilment_market"] = (
    raw_shipments["fulfilment_market"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_shipment_event["fulfilment_country"] = (
    silver_shipment_event["fulfilment_market"]
    .map(country_alias_map)
)

silver_shipment_event["event_timestamp"] = (
    parse_mixed_datetime(
        raw_shipments["event_time"]
    )
)

silver_shipment_event["promised_delivery_date"] = (
    parse_mixed_datetime(
        raw_shipments["promised_date"]
    )
    .dt.normalize()
)

silver_shipment_event["source_file"] = (
    raw_shipments["source_file"]
)

silver_shipment_event["ingested_at"] = (
    parse_mixed_datetime(
        raw_shipments["ingested_at"]
    )
)

print("Shipment events standardised.")
display(silver_shipment_event.head())
```

## Build Silver settlement, payout, and FX tables

Settlement data is standardised for financial reconciliation.

Order-level settlement references are validated where available. Missing references are retained as unallocated settlement items for investigation.

```python
# Helpers for standardising settlement and payout currencies

partner_base_currency_map = (
    partner_ref
    .set_index("partner_id")["base_currency"]
    .to_dict()
)


def standardise_settlement_currency(
    currency_value,
    partner_id
):
    if pd.isna(currency_value):
        return pd.NA

    value = str(currency_value).strip()

    if value in ["GBP", "£", "Pound Sterling"]:
        return "GBP"

    if value in ["EUR", "€", "Euro"]:
        return "EUR"

    if value in ["SEK", "Swedish Krona"]:
        return "SEK"

    if value in ["DKK", "Danish Krone"]:
        return "DKK"

    # "kr" is ambiguous, so use the partner's base currency.
    if value == "kr":
        return partner_base_currency_map.get(
            partner_id,
            pd.NA
        )

    return pd.NA


settlement_line_type_map = {
    "SALE": "sale",
    "REFUND": "refund",
    "ADJUSTMENT": "adjustment"
}
```

```python
# Load and standardise synthetic FX rates

raw_fx = raw_tables[
    "fx_daily_rates_export"
].copy()

silver_fx_rate = pd.DataFrame()

silver_fx_rate["source_row_number"] = (
    raw_fx.index + 1
)

silver_fx_rate["rate_date"] = (
    parse_mixed_datetime(
        raw_fx["rate_date"]
    )
    .dt.normalize()
)

silver_fx_rate["from_currency"] = (
    raw_fx["source_currency"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_fx_rate["to_currency"] = (
    raw_fx["reporting_currency"]
    .astype("string")
    .str.strip()
    .str.upper()
)

silver_fx_rate["rate_to_gbp"] = (
    parse_mixed_number(
        raw_fx["conversion_rate"]
    )
)

silver_fx_rate["rate_source"] = (
    raw_fx["rate_source"]
    .astype("string")
    .str.strip()
)

silver_fx_rate["source_file"] = (
    raw_fx["source_file"]
)

silver_fx_rate["ingested_at"] = (
    parse_mixed_datetime(
        raw_fx["ingested_at"]
    )
)

silver_fx_rate["possible_duplicate_rate"] = (
    silver_fx_rate[
        [
            "rate_date",
            "from_currency",
            "to_currency"
        ]
    ]
    .duplicated(keep=False)
)

fx_reason_rows = []

for _, row in silver_fx_rate.iterrows():
    reasons = []

    if pd.isna(row["rate_date"]):
        reasons.append("invalid_rate_date")

    if pd.isna(row["from_currency"]):
        reasons.append("missing_source_currency")

    if pd.isna(row["to_currency"]):
        reasons.append("missing_reporting_currency")

    if (
        pd.isna(row["rate_to_gbp"])
        or row["rate_to_gbp"] <= 0
    ):
        reasons.append("invalid_fx_rate")

    if row["possible_duplicate_rate"]:
        reasons.append("possible_duplicate_fx_rate")

    fx_reason_rows.append(
        "; ".join(reasons)
        if reasons
        else pd.NA
    )

silver_fx_rate["quality_reasons"] = (
    fx_reason_rows
)

critical_fx_reasons = [
    "invalid_rate_date",
    "missing_source_currency",
    "missing_reporting_currency",
    "invalid_fx_rate"
]

review_fx_reasons = [
    "possible_duplicate_fx_rate"
]

silver_fx_rate["record_status"] = "valid"

for reason in critical_fx_reasons:
    silver_fx_rate.loc[
        silver_fx_rate["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False),
        "record_status"
    ] = "quarantine"

for reason in review_fx_reasons:
    review_mask = (
        silver_fx_rate["quality_reasons"]
        .fillna("")
        .str.contains(reason, regex=False)
        & (
            silver_fx_rate["record_status"]
            == "valid"
        )
    )

    silver_fx_rate.loc[
        review_mask,
        "record_status"
    ] = "review"

silver_fx_rate["analysis_included"] = (
    silver_fx_rate["record_status"]
    != "quarantine"
)

print("FX records by status:")
display(
    silver_fx_rate["record_status"]
    .value_counts()
    .to_frame("record_count")
)
```

## Build Gold analytical tables

The Gold layer contains business-ready tables.

These tables use only records that passed the relevant Silver-layer inclusion rules, while keeping quality flags available for investigation.

```python
# Aggregate valid order lines to the order level

valid_order_lines = silver_order_line[
    silver_order_line["analysis_included"]
].copy()

order_line_totals = (
    valid_order_lines
    .groupby("order_id", as_index=False)
    .agg(
        line_count=("line_id", "nunique"),
        units=("quantity", "sum"),
        line_gross_amount=("line_gross_amount", "sum"),
        line_discount=("line_discount", "sum"),
        line_net_amount=("line_net_amount", "sum")
    )
)

valid_order_headers = silver_order_header[
    silver_order_header["analysis_included"]
].copy()

gold_order_metrics = (
    valid_order_headers[
        [
            "order_id",
            "partner_id",
            "country_code",
            "order_timestamp",
            "order_status",
            "currency_code",
            "shipping_amount",
            "tax_amount"
        ]
    ]
    .merge(
        order_line_totals,
        on="order_id",
        how="left"
    )
)

gold_order_metrics["order_date"] = (
    gold_order_metrics["order_timestamp"]
    .dt.normalize()
)

gold_order_metrics["missing_line_data_flag"] = (
    gold_order_metrics["line_count"].isna()
)

gold_order_metrics["line_count"] = (
    gold_order_metrics["line_count"]
    .fillna(0)
)

gold_order_metrics["units"] = (
    gold_order_metrics["units"]
    .fillna(0)
)

for column in [
    "line_gross_amount",
    "line_discount",
    "line_net_amount"
]:
    gold_order_metrics[column] = (
        gold_order_metrics[column]
        .fillna(0)
    )

gold_order_metrics["shipping_amount"] = (
    gold_order_metrics["shipping_amount"]
    .fillna(0)
)

gold_order_metrics["tax_amount"] = (
    gold_order_metrics["tax_amount"]
    .fillna(0)
)

gold_order_metrics["gross_merchandise_value"] = (
    gold_order_metrics["line_gross_amount"]
)

gold_order_metrics["net_merchandise_value"] = (
    gold_order_metrics["line_net_amount"]
)

gold_order_metrics["net_sales_before_returns"] = (
    gold_order_metrics["net_merchandise_value"]
    + gold_order_metrics["shipping_amount"]
).round(2)

gold_order_metrics["cancelled_flag"] = (
    gold_order_metrics["order_status"]
    == "cancelled"
)

gold_order_metrics["returned_or_refunded_flag"] = (
    gold_order_metrics["order_status"]
    .isin(["returned", "refunded"])
)

print("Gold order metrics created:")
print(gold_order_metrics.shape)

display(
    gold_order_metrics.head()
)
```

```python
# Prepare one FX rate per currency and date

valid_fx = silver_fx_rate[
    silver_fx_rate["analysis_included"]
].copy()

valid_fx = (
    valid_fx
    .sort_values("ingested_at")
    .drop_duplicates(
        subset=[
            "rate_date",
            "from_currency",
            "to_currency"
        ],
        keep="last"
    )
)

fx_for_orders = valid_fx[
    [
        "rate_date",
        "from_currency",
        "to_currency",
        "rate_to_gbp"
    ]
].rename(
    columns={
        "rate_date": "order_date",
        "from_currency": "currency_code"
    }
)

gold_order_metrics = gold_order_metrics.merge(
    fx_for_orders,
    on=[
        "order_date",
        "currency_code"
    ],
    how="left"
)

gold_order_metrics["fx_missing_flag"] = (
    gold_order_metrics["rate_to_gbp"].isna()
)

gold_order_metrics["rate_to_gbp"] = (
    gold_order_metrics["rate_to_gbp"]
    .fillna(1.0)
)

for column in [
    "gross_merchandise_value",
    "net_merchandise_value",
    "net_sales_before_returns",
    "shipping_amount",
    "tax_amount"
]:
    gold_order_metrics[
        f"{column}_gbp"
    ] = (
        gold_order_metrics[column]
        * gold_order_metrics["rate_to_gbp"]
    ).round(2)

print("Order values converted to GBP.")
display(
    gold_order_metrics[
        [
            "order_id",
            "currency_code",
            "rate_to_gbp",
            "net_sales_before_returns",
            "net_sales_before_returns_gbp",
            "fx_missing_flag"
        ]
    ].head()
)
```

```python
# Aggregate order metrics by partner, country, and day

gold_partner_daily_metrics = (
    gold_order_metrics
    .groupby(
        [
            "order_date",
            "partner_id",
            "country_code"
        ],
        as_index=False
    )
    .agg(
        order_count=("order_id", "nunique"),
        units=("units", "sum"),
        gross_merchandise_value_gbp=(
            "gross_merchandise_value_gbp",
            "sum"
        ),
        net_merchandise_value_gbp=(
            "net_merchandise_value_gbp",
            "sum"
        ),
        net_sales_before_returns_gbp=(
            "net_sales_before_returns_gbp",
            "sum"
        ),
        cancelled_order_count=(
            "cancelled_flag",
            "sum"
        ),
        returned_or_refunded_order_count=(
            "returned_or_refunded_flag",
            "sum"
        ),
        orders_missing_line_data=(
            "missing_line_data_flag",
            "sum"
        ),
        orders_missing_fx=(
            "fx_missing_flag",
            "sum"
        )
    )
)

gold_partner_daily_metrics["average_order_value_gbp"] = (
    gold_partner_daily_metrics[
        "net_sales_before_returns_gbp"
    ]
    / gold_partner_daily_metrics["order_count"]
).round(2)

gold_partner_daily_metrics["cancellation_rate"] = (
    gold_partner_daily_metrics[
        "cancelled_order_count"
    ]
    / gold_partner_daily_metrics["order_count"]
).round(4)

gold_partner_daily_metrics["return_or_refund_rate"] = (
    gold_partner_daily_metrics[
        "returned_or_refunded_order_count"
    ]
    / gold_partner_daily_metrics["order_count"]
).round(4)

print("Partner daily metrics created.")
display(gold_partner_daily_metrics.head())
```

## Reconcile expected and reported settlement amounts

This table compares an explainable expected settlement amount with the amount reported by the fictional partner.

Differences are prioritised for investigation rather than automatically labelled as errors.

```python
# Prepare FX rates for settlement-date conversion

valid_fx_for_settlement = silver_fx_rate[
    silver_fx_rate["analysis_included"]
].copy()

valid_fx_for_settlement = (
    valid_fx_for_settlement
    .sort_values(
        [
            "from_currency",
            "rate_date",
            "ingested_at"
        ]
    )
    .drop_duplicates(
        subset=[
            "from_currency",
            "to_currency",
            "rate_date"
        ],
        keep="last"
    )
)

latest_fx_rate_by_currency = (
    valid_fx_for_settlement
    .sort_values("rate_date")
    .groupby("from_currency")
    .tail(1)
    .set_index("from_currency")["rate_to_gbp"]
    .to_dict()
)

settlement_for_reconciliation = (
    silver_settlement_line[
        silver_settlement_line["analysis_included"]
    ]
    .copy()
)

settlement_for_reconciliation["payout_date_only"] = (
    settlement_for_reconciliation["payout_date"]
    .dt.normalize()
)

settlement_for_reconciliation = (
    settlement_for_reconciliation
    .merge(
        valid_fx_for_settlement[
            [
                "rate_date",
                "from_currency",
                "to_currency",
                "rate_to_gbp"
            ]
        ].rename(
            columns={
                "rate_date": "payout_date_only",
                "from_currency": "currency_code"
            }
        ),
        on=[
            "payout_date_only",
            "currency_code"
        ],
        how="left"
    )
)

settlement_for_reconciliation["fx_missing_flag"] = (
    settlement_for_reconciliation["rate_to_gbp"]
    .isna()
)

settlement_for_reconciliation["rate_to_gbp"] = (
    settlement_for_reconciliation["rate_to_gbp"]
    .fillna(
        settlement_for_reconciliation[
            "currency_code"
        ].map(latest_fx_rate_by_currency)
    )
    .fillna(1.0)
)

settlement_for_reconciliation["net_amount_gbp"] = (
    settlement_for_reconciliation["net_amount"]
    * settlement_for_reconciliation["rate_to_gbp"]
).round(2)

print("Settlement amounts converted to GBP.")
display(
    settlement_for_reconciliation[
        [
            "settlement_line_id",
            "partner_id",
            "currency_code",
            "net_amount",
            "rate_to_gbp",
            "net_amount_gbp",
            "fx_missing_flag"
        ]
    ].head()
)
```

```python
# Aggregate reported settlement amounts by order

reported_settlement_by_order = (
    settlement_for_reconciliation[
        settlement_for_reconciliation[
            "order_reference"
        ].notna()
    ]
    .groupby(
        "order_reference",
        as_index=False
    )
    .agg(
        reported_settlement_gbp=(
            "net_amount_gbp",
            "sum"
        ),
        settlement_line_count=(
            "settlement_line_id",
            "nunique"
        ),
        settlement_fx_missing_count=(
            "fx_missing_flag",
            "sum"
        )
    )
    .rename(
        columns={
            "order_reference": "order_id"
        }
    )
)

print(
    "Orders with reported settlement:",
    f"{len(reported_settlement_by_order):,}"
)

display(
    reported_settlement_by_order.head()
)
```

```python
# Create an explainable expected settlement amount per order

expected_settlement = gold_order_metrics[
    gold_order_metrics["order_status"] != "cancelled"
].copy()

expected_settlement["expected_gross_settlement_gbp"] = (
    expected_settlement["net_merchandise_value_gbp"]
    + expected_settlement["shipping_amount_gbp"]
).round(2)

expected_settlement["expected_platform_fee_gbp"] = (
    expected_settlement["expected_gross_settlement_gbp"]
    * expected_settlement["partner_id"].map(
        partner_fee_rates
    )
).round(2)

# Use a documented synthetic payment-fee assumption
expected_settlement["expected_payment_fee_gbp"] = (
    expected_settlement["expected_gross_settlement_gbp"]
    * 0.025
).round(2)

expected_settlement["expected_tax_withheld_gbp"] = (
    expected_settlement["tax_amount_gbp"]
).round(2)

expected_settlement["expected_sale_settlement_gbp"] = (
    expected_settlement["expected_gross_settlement_gbp"]
    - expected_settlement["expected_platform_fee_gbp"]
    - expected_settlement["expected_payment_fee_gbp"]
    - expected_settlement["expected_tax_withheld_gbp"]
).round(2)

print("Expected order settlements created.")

display(
    expected_settlement[
        [
            "order_id",
            "partner_id",
            "expected_gross_settlement_gbp",
            "expected_platform_fee_gbp",
            "expected_payment_fee_gbp",
            "expected_tax_withheld_gbp",
            "expected_sale_settlement_gbp"
        ]
    ].head()
)
```

```python
# Convert return refunds to GBP and aggregate by order

valid_returns_for_reconciliation = silver_return_case[
    silver_return_case["analysis_included"]
].copy()

valid_returns_for_reconciliation["received_date_only"] = (
    valid_returns_for_reconciliation["received_at"]
    .dt.normalize()
)

valid_returns_for_reconciliation = (
    valid_returns_for_reconciliation
    .merge(
        valid_fx_for_settlement[
            [
                "rate_date",
                "from_currency",
                "to_currency",
                "rate_to_gbp"
            ]
        ].rename(
            columns={
                "rate_date": "received_date_only",
                "from_currency": "currency_code"
            }
        ),
        on=[
            "received_date_only",
            "currency_code"
        ],
        how="left"
    )
)

valid_returns_for_reconciliation["rate_to_gbp"] = (
    valid_returns_for_reconciliation["rate_to_gbp"]
    .fillna(
        valid_returns_for_reconciliation[
            "currency_code"
        ].map(latest_fx_rate_by_currency)
    )
    .fillna(1.0)
)

valid_returns_for_reconciliation["refund_amount_gbp"] = (
    valid_returns_for_reconciliation["refund_amount"]
    * valid_returns_for_reconciliation["rate_to_gbp"]
).round(2)

refunds_by_order = (
    valid_returns_for_reconciliation
    .groupby("order_id", as_index=False)
    .agg(
        expected_refund_amount_gbp=(
            "refund_amount_gbp",
            "sum"
        ),
        return_case_count=(
            "return_id",
            "nunique"
        )
    )
)

expected_settlement = expected_settlement.merge(
    refunds_by_order,
    on="order_id",
    how="left"
)

expected_settlement[
    "expected_refund_amount_gbp"
] = (
    expected_settlement[
        "expected_refund_amount_gbp"
    ]
    .fillna(0)
)

expected_settlement[
    "expected_net_settlement_gbp"
] = (
    expected_settlement[
        "expected_sale_settlement_gbp"
    ]
    - expected_settlement[
        "expected_refund_amount_gbp"
    ]
).round(2)

print("Expected refunds included.")
display(
    expected_settlement[
        [
            "order_id",
            "expected_sale_settlement_gbp",
            "expected_refund_amount_gbp",
            "expected_net_settlement_gbp"
        ]
    ].head()
)
```

```python
# Compare expected settlement with reported settlement

gold_settlement_reconciliation = (
    expected_settlement[
        [
            "order_id",
            "partner_id",
            "country_code",
            "order_date",
            "order_status",
            "expected_gross_settlement_gbp",
            "expected_platform_fee_gbp",
            "expected_payment_fee_gbp",
            "expected_tax_withheld_gbp",
            "expected_sale_settlement_gbp",
            "expected_refund_amount_gbp",
            "expected_net_settlement_gbp",
            "return_case_count"
        ]
    ]
    .merge(
        reported_settlement_by_order,
        on="order_id",
        how="left"
    )
)

gold_settlement_reconciliation[
    "reported_settlement_gbp"
] = (
    gold_settlement_reconciliation[
        "reported_settlement_gbp"
    ]
    .fillna(0)
)

gold_settlement_reconciliation[
    "settlement_line_count"
] = (
    gold_settlement_reconciliation[
        "settlement_line_count"
    ]
    .fillna(0)
)

gold_settlement_reconciliation[
    "settlement_fx_missing_count"
] = (
    gold_settlement_reconciliation[
        "settlement_fx_missing_count"
    ]
    .fillna(0)
)

gold_settlement_reconciliation[
    "settlement_variance_gbp"
] = (
    gold_settlement_reconciliation[
        "reported_settlement_gbp"
    ]
    - gold_settlement_reconciliation[
        "expected_net_settlement_gbp"
    ]
).round(2)

gold_settlement_reconciliation[
    "absolute_variance_gbp"
] = (
    gold_settlement_reconciliation[
        "settlement_variance_gbp"
    ]
    .abs()
)

gold_settlement_reconciliation[
    "settlement_present_flag"
] = (
    gold_settlement_reconciliation[
        "settlement_line_count"
    ]
    > 0
)

gold_settlement_reconciliation[
    "reconciliation_status"
] = np.select(
    [
        (
            ~gold_settlement_reconciliation[
                "settlement_present_flag"
            ]
            & gold_settlement_reconciliation[
                "order_status"
            ].eq("pending")
        ),
        gold_settlement_reconciliation[
            "absolute_variance_gbp"
        ] <= 1.00,
        gold_settlement_reconciliation[
            "absolute_variance_gbp"
        ] <= 5.00,
        ~gold_settlement_reconciliation[
            "settlement_present_flag"
        ]
    ],
    [
        "not_yet_settled",
        "within_tolerance",
        "minor_variance",
        "missing_reported_settlement"
    ],
    default="investigate"
)

gold_settlement_reconciliation[
    "commercial_priority"
] = np.select(
    [
        gold_settlement_reconciliation[
            "absolute_variance_gbp"
        ] >= 100,
        gold_settlement_reconciliation[
            "absolute_variance_gbp"
        ] >= 25
    ],
    [
        "high",
        "medium"
    ],
    default="low"
)

print("Settlement reconciliation created.")
display(
    gold_settlement_reconciliation[
        "reconciliation_status"
    ]
    .value_counts()
    .to_frame("order_count")
)
```

```python
# Summarise settlement lines without an order-level reference

unallocated_settlement = (
    settlement_for_reconciliation[
        settlement_for_reconciliation[
            "unallocated_order_flag"
        ]
    ]
    .groupby(
        [
            "partner_id",
            "currency_code",
            "line_type",
            "payout_date_only"
        ],
        as_index=False
    )
    .agg(
        unallocated_line_count=(
            "settlement_line_id",
            "nunique"
        ),
        unallocated_amount_gbp=(
            "net_amount_gbp",
            "sum"
        ),
        absolute_unallocated_amount_gbp=(
            "net_amount_gbp",
            lambda values: values.abs().sum()
        )
    )
)

unallocated_settlement[
    "commercial_priority"
] = np.select(
    [
        unallocated_settlement[
            "absolute_unallocated_amount_gbp"
        ] >= 100,
        unallocated_settlement[
            "absolute_unallocated_amount_gbp"
        ] >= 25
    ],
    [
        "high",
        "medium"
    ],
    default="low"
)

print("Unallocated settlement summary created.")
display(
    unallocated_settlement.head()
)
```

## Detect anomalies and build the exception queue

The exception queue combines commercial anomalies and data-quality issues.

Anomaly detection uses rolling historical baselines. Every flagged item includes evidence, commercial impact, and a recommended next action.

```python
# Calculate rolling partner and country baselines

metric_baseline_parts = []

metric_group_columns = [
    "partner_id",
    "country_code"
]

for group_key, group in gold_partner_daily_metrics.groupby(
    metric_group_columns,
    dropna=False
):
    group = group.sort_values(
        "order_date"
    ).copy()

    sales_series = (
        group["net_sales_before_returns_gbp"]
        .astype(float)
    )

    order_series = (
        group["order_count"]
        .astype(float)
    )

    group["sales_baseline_gbp"] = (
        sales_series
        .shift(1)
        .rolling(
            window=28,
            min_periods=7
        )
        .median()
    )

    group["sales_baseline_std_gbp"] = (
        sales_series
        .shift(1)
        .rolling(
            window=28,
            min_periods=7
        )
        .std()
    )

    group["orders_baseline"] = (
        order_series
        .shift(1)
        .rolling(
            window=28,
            min_periods=7
        )
        .median()
    )

    group["sales_deviation_gbp"] = (
        group["net_sales_before_returns_gbp"]
        - group["sales_baseline_gbp"]
    ).round(2)

    group["sales_deviation_pct"] = (
        group["sales_deviation_gbp"]
        / group["sales_baseline_gbp"]
        .replace(0, np.nan)
    ).round(4)

    group["sales_z_score"] = (
        group["sales_deviation_gbp"]
        / group["sales_baseline_std_gbp"]
        .replace(0, np.nan)
    ).round(2)

    group["orders_deviation_pct"] = (
        (
            group["order_count"]
            - group["orders_baseline"]
        )
        / group["orders_baseline"]
        .replace(0, np.nan)
    ).round(4)

    metric_baseline_parts.append(group)

gold_partner_metric_anomalies = pd.concat(
    metric_baseline_parts,
    ignore_index=True
)

print(
    "Baseline metrics created:",
    gold_partner_metric_anomalies.shape
)

display(
    gold_partner_metric_anomalies[
        [
            "order_date",
            "partner_id",
            "country_code",
            "net_sales_before_returns_gbp",
            "sales_baseline_gbp",
            "sales_deviation_pct",
            "sales_z_score"
        ]
    ].tail()
)
```

```python
# Classify explainable partner metric anomalies

gold_partner_metric_anomalies["metric_anomaly_reason"] = pd.NA

low_sales_mask = (
    gold_partner_metric_anomalies[
        "sales_deviation_pct"
    ] <= -0.40
) & (
    gold_partner_metric_anomalies[
        "sales_baseline_gbp"
    ] > 0
)

high_sales_mask = (
    gold_partner_metric_anomalies[
        "sales_deviation_pct"
    ] >= 0.80
) & (
    gold_partner_metric_anomalies[
        "sales_baseline_gbp"
    ] > 0
)

large_z_score_mask = (
    gold_partner_metric_anomalies[
        "sales_z_score"
    ].abs() >= 3
)

high_cancellation_mask = (
    gold_partner_metric_anomalies[
        "cancellation_rate"
    ] >= 0.15
)

metric_anomaly_reason = np.select(
    [
        low_sales_mask & large_z_score_mask,
        high_sales_mask & large_z_score_mask,
        low_sales_mask,
        high_sales_mask,
        high_cancellation_mask
    ],
    [
        "sales_significantly_below_baseline",
        "sales_significantly_above_baseline",
        "sales_below_baseline",
        "sales_above_baseline",
        "high_cancellation_rate"
    ],
    default=None
)

gold_partner_metric_anomalies[
    "metric_anomaly_reason"
] = metric_anomaly_reason

gold_partner_metric_anomalies["metric_anomaly_flag"] = (
    gold_partner_metric_anomalies[
        "metric_anomaly_reason"
    ].notna()
)

gold_partner_metric_anomalies[
    "estimated_metric_impact_gbp"
] = (
    gold_partner_metric_anomalies[
        "sales_deviation_gbp"
    ].abs()
)

print("Metric anomalies detected:")
display(
    gold_partner_metric_anomalies[
        "metric_anomaly_reason"
    ]
    .value_counts(dropna=False)
    .to_frame("row_count")
)
```

```python
# Create listing and inventory exception records

listing_exception_base = listing_health.copy()

listing_exception_base["currency_rate_to_gbp"] = (
    listing_exception_base["currency_code"]
    .map(latest_fx_rate_by_currency)
    .fillna(1.0)
)

listing_exception_base[
    "estimated_inventory_value_at_risk_gbp"
] = (
    listing_exception_base[
        "inventory_available_units"
    ]
    .fillna(0)
    .clip(lower=0)
    * listing_exception_base[
        "price_local"
    ].fillna(0)
    * listing_exception_base[
        "currency_rate_to_gbp"
    ]
).round(2)

listing_exception_frames = []

offline_records = listing_exception_base[
    listing_exception_base["listing_offline_flag"]
].copy()

offline_records["issue_type"] = (
    "listing_offline_or_paused"
)

offline_records["evidence"] = (
    "Listing status is offline or paused"
)

offline_records["recommended_action"] = (
    "Investigate source status, content requirements, and inventory availability"
)

listing_exception_frames.append(
    offline_records
)

inventory_disagreement_records = listing_exception_base[
    listing_exception_base[
        "listing_inventory_disagreement_flag"
    ]
].copy()

inventory_disagreement_records["issue_type"] = (
    "listing_inventory_disagreement"
)

inventory_disagreement_records["evidence"] = (
    "Listing stock differs materially from latest inventory stock"
)

inventory_disagreement_records["recommended_action"] = (
    "Compare feed timestamps, reservations, and partner inventory rules"
)

listing_exception_frames.append(
    inventory_disagreement_records
)

stale_inventory_records = listing_exception_base[
    listing_exception_base["inventory_stale_flag"]
].copy()

stale_inventory_records["issue_type"] = (
    "stale_inventory"
)

stale_inventory_records["evidence"] = (
    "Latest inventory update is more than 48 hours from the listing snapshot"
)

stale_inventory_records["recommended_action"] = (
    "Investigate feed freshness and confirm the expected update frequency"
)

listing_exception_frames.append(
    stale_inventory_records
)

incomplete_content_records = listing_exception_base[
    listing_exception_base["incomplete_content_flag"]
].copy()

incomplete_content_records["issue_type"] = (
    "incomplete_listing_content"
)

incomplete_content_records["evidence"] = (
    "One or more required listing-content fields are missing"
)

incomplete_content_records["recommended_action"] = (
    "Review missing title, category, image, or other required attributes"
)

listing_exception_frames.append(
    incomplete_content_records
)

listing_exception_records = pd.concat(
    listing_exception_frames,
    ignore_index=True
)

print(
    "Listing exception records:",
    f"{len(listing_exception_records):,}"
)

display(
    listing_exception_records[
        "issue_type"
    ]
    .value_counts()
    .to_frame("record_count")
)
```

```python
# Create settlement reconciliation exception records

settlement_exception_records = (
    gold_settlement_reconciliation[
        gold_settlement_reconciliation[
            "reconciliation_status"
        ].isin(
            [
                "investigate",
                "missing_reported_settlement"
            ]
        )
    ]
    .copy()
)

settlement_exception_records["issue_type"] = (
    "settlement_variance"
)

settlement_exception_records["evidence"] = (
    settlement_exception_records[
        "reconciliation_status"
    ]
    + " with variance of GBP "
    + settlement_exception_records[
        "settlement_variance_gbp"
    ]
    .round(2)
    .astype(str)
)

settlement_exception_records["recommended_action"] = (
    "Review settlement line types, fees, refunds, FX, timing, and order references"
)

settlement_exception_records[
    "commercial_impact_gbp"
] = (
    settlement_exception_records[
        "absolute_variance_gbp"
    ]
)

unallocated_exception_records = (
    unallocated_settlement[
        unallocated_settlement[
            "absolute_unallocated_amount_gbp"
        ] > 0
    ]
    .copy()
)

unallocated_exception_records["issue_type"] = (
    "unallocated_settlement"
)

unallocated_exception_records["evidence"] = (
    "Settlement amount has no direct order-level reference"
)

unallocated_exception_records["recommended_action"] = (
    "Investigate statement or payout-batch allocation rules"
)

unallocated_exception_records[
    "commercial_impact_gbp"
] = (
    unallocated_exception_records[
        "absolute_unallocated_amount_gbp"
    ]
)

print(
    "Settlement exception records:",
    f"{len(settlement_exception_records):,}"
)

print(
    "Unallocated settlement records:",
    f"{len(unallocated_exception_records):,}"
)
```

```python
# Create a common exception-queue structure

exception_frames = []

# Listing exceptions
if len(listing_exception_records) > 0:
    listing_queue = pd.DataFrame({
        "source_table": "gold_listing_health",
        "issue_type": listing_exception_records[
            "issue_type"
        ],
        "partner_id": listing_exception_records[
            "partner_id"
        ],
        "country_code": listing_exception_records[
            "country_code"
        ],
        "internal_sku": listing_exception_records[
            "internal_sku"
        ],
        "order_id": pd.NA,
        "event_date": listing_exception_records[
            "snapshot_date"
        ],
        "commercial_impact_gbp": listing_exception_records[
            "estimated_inventory_value_at_risk_gbp"
        ],
        "evidence": listing_exception_records[
            "evidence"
        ],
        "recommended_action": listing_exception_records[
            "recommended_action"
        ]
    })

    exception_frames.append(listing_queue)

# Settlement exceptions
if len(settlement_exception_records) > 0:
    settlement_queue = pd.DataFrame({
        "source_table": "gold_settlement_reconciliation",
        "issue_type": settlement_exception_records[
            "issue_type"
        ],
        "partner_id": settlement_exception_records[
            "partner_id"
        ],
        "country_code": settlement_exception_records[
            "country_code"
        ],
        "internal_sku": pd.NA,
        "order_id": settlement_exception_records[
            "order_id"
        ],
        "event_date": settlement_exception_records[
            "order_date"
        ],
        "commercial_impact_gbp": settlement_exception_records[
            "commercial_impact_gbp"
        ],
        "evidence": settlement_exception_records[
            "evidence"
        ],
        "recommended_action": settlement_exception_records[
            "recommended_action"
        ]
    })

    exception_frames.append(settlement_queue)

# Unallocated settlement exceptions
if len(unallocated_exception_records) > 0:
    unallocated_queue = pd.DataFrame({
        "source_table": "gold_unallocated_settlement",
        "issue_type": unallocated_exception_records[
            "issue_type"
        ],
        "partner_id": unallocated_exception_records[
            "partner_id"
        ],
        "country_code": pd.NA,
        "internal_sku": pd.NA,
        "order_id": pd.NA,
        "event_date": unallocated_exception_records[
            "payout_date_only"
        ],
        "commercial_impact_gbp": unallocated_exception_records[
            "commercial_impact_gbp"
        ],
        "evidence": unallocated_exception_records[
            "evidence"
        ],
        "recommended_action": unallocated_exception_records[
            "recommended_action"
        ]
    })

    exception_frames.append(unallocated_queue)

# Partner metric anomalies
metric_anomaly_records = (
    gold_partner_metric_anomalies[
        gold_partner_metric_anomalies[
            "metric_anomaly_flag"
        ]
    ]
    .copy()
)

if len(metric_anomaly_records) > 0:
    metric_queue = pd.DataFrame({
        "source_table": "gold_partner_daily_metrics",
        "issue_type": "partner_metric_anomaly",
        "partner_id": metric_anomaly_records[
            "partner_id"
        ],
        "country_code": metric_anomaly_records[
            "country_code"
        ],
        "internal_sku": pd.NA,
        "order_id": pd.NA,
        "event_date": metric_anomaly_records[
            "order_date"
        ],
        "commercial_impact_gbp": metric_anomaly_records[
            "estimated_metric_impact_gbp"
        ],
        "evidence": (
            metric_anomaly_records[
                "metric_anomaly_reason"
            ]
            + "; observed sales GBP "
            + metric_anomaly_records[
                "net_sales_before_returns_gbp"
            ]
            .round(2)
            .astype(str)
            + "; baseline sales GBP "
            + metric_anomaly_records[
                "sales_baseline_gbp"
            ]
            .round(2)
            .astype(str)
        ),
        "recommended_action": (
            "Review source completeness, partner activity, promotions, and operational events"
        )
    })

    exception_frames.append(metric_queue)

# Standardise exception-frame columns before concatenation

exception_queue_columns = [
    "source_table",
    "issue_type",
    "partner_id",
    "country_code",
    "internal_sku",
    "order_id",
    "event_date",
    "commercial_impact_gbp",
    "evidence",
    "recommended_action"
]

string_exception_columns = [
    "source_table",
    "issue_type",
    "partner_id",
    "country_code",
    "internal_sku",
    "order_id",
    "evidence",
    "recommended_action"
]

normalised_exception_frames = []

for frame in exception_frames:
    frame = frame.reindex(
        columns=exception_queue_columns
    ).copy()

    for column in string_exception_columns:
        frame[column] = frame[column].astype("string")

    frame["commercial_impact_gbp"] = pd.to_numeric(
        frame["commercial_impact_gbp"],
        errors="coerce"
    ).astype("float64")

    normalised_exception_frames.append(frame)

gold_exception_queue = pd.concat(
    normalised_exception_frames,
    ignore_index=True
)

gold_exception_queue["commercial_impact_gbp"] = (
    pd.to_numeric(
        gold_exception_queue[
            "commercial_impact_gbp"
        ],
        errors="coerce"
    )
    .fillna(0)
    .round(2)
)

impact_q75 = gold_exception_queue[
    "commercial_impact_gbp"
].quantile(0.75)

impact_q95 = gold_exception_queue[
    "commercial_impact_gbp"
].quantile(0.95)

gold_exception_queue["severity"] = np.select(
    [
        gold_exception_queue[
            "commercial_impact_gbp"
        ] >= impact_q95,
        gold_exception_queue[
            "commercial_impact_gbp"
        ] >= impact_q75
    ],
    [
        "high",
        "medium"
    ],
    default="low"
)

gold_exception_queue["review_status"] = (
    "open"
)

gold_exception_queue["review_comment"] = (
    pd.NA
)

gold_exception_queue["exception_id"] = [
    f"EXC-{number:07d}"
    for number in range(
        1,
        len(gold_exception_queue) + 1
    )
]

gold_exception_queue = gold_exception_queue[
    [
        "exception_id",
        "source_table",
        "issue_type",
        "partner_id",
        "country_code",
        "internal_sku",
        "order_id",
        "event_date",
        "severity",
        "commercial_impact_gbp",
        "evidence",
        "recommended_action",
        "review_status",
        "review_comment"
    ]
]

print(
    "Exception queue rows:",
    f"{len(gold_exception_queue):,}"
)

display(
    gold_exception_queue.head()
)
```

## Create prioritised review and data-quality outputs

The full exception queue is retained for auditability.

A separate prioritised queue ranks exceptions by severity and commercial impact so that an analyst can focus on the most important items first.

```python
# Summarise exceptions by issue type, partner, and severity

gold_exception_summary = (
    gold_exception_queue
    .groupby(
        [
            "source_table",
            "issue_type",
            "partner_id",
            "severity"
        ],
        dropna=False,
        as_index=False
    )
    .agg(
        exception_count=(
            "exception_id",
            "nunique"
        ),
        total_commercial_impact_gbp=(
            "commercial_impact_gbp",
            "sum"
        ),
        maximum_commercial_impact_gbp=(
            "commercial_impact_gbp",
            "max"
        ),
        open_exception_count=(
            "review_status",
            lambda values: (values == "open").sum()
        )
    )
)

gold_exception_summary[
    "total_commercial_impact_gbp"
] = (
    gold_exception_summary[
        "total_commercial_impact_gbp"
    ]
    .round(2)
)

gold_exception_summary[
    "maximum_commercial_impact_gbp"
] = (
    gold_exception_summary[
        "maximum_commercial_impact_gbp"
    ]
    .round(2)
)

print("Exception summary created:")
display(
    gold_exception_summary
    .sort_values(
        "total_commercial_impact_gbp",
        ascending=False
    )
    .head(20)
)
```

```python
# Create a ranked exception queue for operational review

severity_score_map = {
    "high": 3,
    "medium": 2,
    "low": 1
}

prioritised_queue = gold_exception_queue.copy()

prioritised_queue["severity_score"] = (
    prioritised_queue["severity"]
    .map(severity_score_map)
    .fillna(0)
)

prioritised_queue["commercial_impact_log"] = (
    np.log1p(
        prioritised_queue[
            "commercial_impact_gbp"
        ].clip(lower=0)
    )
)

prioritised_queue["priority_score"] = (
    prioritised_queue["severity_score"] * 10
    + prioritised_queue["commercial_impact_log"]
).round(4)

prioritised_queue = (
    prioritised_queue
    .sort_values(
        [
            "priority_score",
            "commercial_impact_gbp"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

prioritised_queue["dashboard_rank"] = (
    np.arange(len(prioritised_queue))
    + 1
)

# Keep the full ranked queue available, but create a compact
# dashboard queue for faster review.
gold_prioritised_exception_queue = (
    prioritised_queue
    .head(10000)
    .copy()
)

gold_prioritised_exception_queue = (
    gold_prioritised_exception_queue[
        [
            "dashboard_rank",
            "exception_id",
            "source_table",
            "issue_type",
            "partner_id",
            "country_code",
            "internal_sku",
            "order_id",
            "event_date",
            "severity",
            "commercial_impact_gbp",
            "evidence",
            "recommended_action",
            "review_status",
            "review_comment",
            "priority_score"
        ]
    ]
)

print(
    "Full exception queue:",
    f"{len(gold_exception_queue):,}"
)

print(
    "Dashboard exception queue:",
    f"{len(gold_prioritised_exception_queue):,}"
)

display(
    gold_prioritised_exception_queue.head(20)
)
```

```python
# Create a data-quality summary across all Silver tables

silver_tables_for_quality = {
    "silver_listing": silver_listing,
    "silver_inventory": silver_inventory,
    "silver_order_header": silver_order_header,
    "silver_order_line": silver_order_line,
    "silver_shipment_event": silver_shipment_event,
    "silver_return_case": silver_return_case,
    "silver_settlement_line": silver_settlement_line,
    "silver_payout_batch": silver_payout_batch,
    "silver_fx_rate": silver_fx_rate
}

quality_summary_rows = []

for table_name, table in silver_tables_for_quality.items():
    row_count = len(table)

    if "record_status" in table.columns:
        valid_count = int(
            (table["record_status"] == "valid").sum()
        )

        review_count = int(
            (table["record_status"] == "review").sum()
        )

        quarantine_count = int(
            (table["record_status"] == "quarantine").sum()
        )

        analysis_included_count = int(
            table["analysis_included"].sum()
        )
    else:
        valid_count = row_count
        review_count = 0
        quarantine_count = 0
        analysis_included_count = row_count

    quality_summary_rows.append({
        "table_name": table_name,
        "row_count": row_count,
        "valid_count": valid_count,
        "review_count": review_count,
        "quarantine_count": quarantine_count,
        "analysis_included_count": analysis_included_count,
        "valid_percentage": round(
            valid_count / row_count * 100,
            2
        ) if row_count else 0,
        "review_percentage": round(
            review_count / row_count * 100,
            2
        ) if row_count else 0,
        "quarantine_percentage": round(
            quarantine_count / row_count * 100,
            2
        ) if row_count else 0
    })

data_quality_summary = pd.DataFrame(
    quality_summary_rows
)

display(data_quality_summary)
```

```python
# Add overall data-quality indicators

overall_quality_summary = pd.DataFrame([
    {
        "metric": "raw_table_count",
        "value": len(raw_tables)
    },
    {
        "metric": "full_exception_queue_rows",
        "value": len(gold_exception_queue)
    },
    {
        "metric": "dashboard_exception_queue_rows",
        "value": len(
            gold_prioritised_exception_queue
        )
    },
    {
        "metric": "high_severity_exception_count",
        "value": int(
            (
                gold_exception_queue["severity"]
                == "high"
            ).sum()
        )
    },
    {
        "metric": "medium_severity_exception_count",
        "value": int(
            (
                gold_exception_queue["severity"]
                == "medium"
            ).sum()
        )
    },
    {
        "metric": "open_exception_count",
        "value": int(
            (
                gold_exception_queue["review_status"]
                == "open"
            ).sum()
        )
    },
    {
        "metric": "total_exception_impact_gbp",
        "value": round(
            gold_exception_queue[
                "commercial_impact_gbp"
            ].sum(),
            2
        )
    }
])

print("Overall quality indicators:")
display(overall_quality_summary)
```
